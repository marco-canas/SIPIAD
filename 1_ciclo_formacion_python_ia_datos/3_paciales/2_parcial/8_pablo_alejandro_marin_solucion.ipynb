
### 2. Definición Estadística de la Mediana

La **mediana** es una medida de tendencia central que representa el valor central de un conjunto de datos numéricos cuando estos se encuentran ordenados de menor a mayor.

Dada una lista de $N$ números ordenados:

* **Si el número de elementos ($N$) es IMPAR:** La mediana es el valor que se encuentra exactamente en la posición central:

$$\text{Mediana} = \text{datos}\left[\frac{N - 1}{2}\right]$$


* **Si el número de elementos ($N$) es PAR:** La mediana es el promedio de los dos valores centrales:

$$\text{Mediana} = \frac{\text{datos}\left[\frac{N}{2} - 1\right] + \text{datos}\left[\frac{N}{2}\right]}{2}$$

### **Punto 1: Análisis de Números Abundantes (Teoría de Números y Condicionales)**

En teoría de números, un número entero positivo $n$ se considera **abundante** si la suma de sus divisores propios (todos los divisores positivos de $n$ excluyendo al propio $n$) es estrictamente **mayor** que $n$.

Por ejemplo, para $n = 12$:

* Divisores propios: $1, 2, 3, 4, 6$.
* Suma de divisores propios: $1 + 2 + 3 + 4 + 6 = 16$.
* Como $16 > 12$, el número $12$ es **abundante**.

#### **Enunciado:**

* Retorne `"Abundante"` si la suma de los divisores propios es mayor que $n$.
* Retorne `"Perfecto"` si la suma de los divisores propios es exactamente igual a $n$.
* Retorne `"Deficiente"` si la suma de los divisores propios es menor que $n$.



```python

```

````python
#1
def evaluar_numero(n):
    # 1. Calcular la suma de todos los divisores propios mediante un ciclo for simple
    suma_divisores = 0
    for i in range(1, n):
        if n % i == 0:
            suma_divisores += i

    # 2. Utilizar estructuras condicionales (if, elif, else) para retornar el resultado
    if suma_divisores > n:
        return "Abundante"
    elif suma_divisores == n:
        return "Perfecto"
    else:
        return "Deficiente"


# --- Pruebas sugeridas ---
print(evaluar_numero(12))  # Debe retornar 'Abundante'
print(evaluar_numero(6))   # Debe retornar 'Perfecto'
print(evaluar_numero(8))   # Debe retornar 'Deficiente'
````


### **Punto 2: Limpieza y Filtro de Datos de Mediciones (Ciencia de Datos y Ciclos `while`)**

Un equipo de sensores ambientales ha recolectado una serie temporal de mediciones numéricas. Sin embargo, debido a fallas de sensor, se han registrado valores negativos no válidos y un valor de parada explícito ($-999$) que indica el fin de la transmisión de datos.

#### **Enunciado:**

2. El proceso debe detenerse inmediatamente cuando se encuentre con el valor de parada `-999` o cuando se llegue al final de la lista.
3. Durante la iteración, utiliza condicionales para ignorar los valores negativos y conservar únicamente las mediciones **mayores o iguales a cero**.
4. La función debe retornar una **nueva lista** con las mediciones válidas recolectadas antes de la señal de parada.

```python

    # Escribe tu código aquí
    pass

# Prueba sugerida:
# datos_sensor = [14.5, 18.2, -3.1, 22.0, -1.0, 19.8, -999, 15.0, 30.0]
# print(filtrar_mediciones(datos_sensor)) # Debe retornar [14.5, 18.2, 22.0, 19.8]

```

---


````python
#2
def filtrar_mediciones(lista_mediciones):
    mediciones_validas = []
    i = 0
    N = len(lista_mediciones)

    # 1. Ciclo while para iterar mientras no lleguemos al final de la lista
    while i < N:
        valor = lista_mediciones[i]

        # 2. Detenerse inmediatamente al encontrar el valor de parada -999
        if valor == -999:
            break

        # 3. Conservar únicamente mediciones mayores o iguales a cero (ignorar negativos)
        if valor >= 0:
            mediciones_validas.append(valor)

        i += 1

    # 4. Retornar la nueva lista con las mediciones válidas recolectadas
    return mediciones_validas


# --- Prueba sugerida ---
datos_sensor = [14.5, 18.2, -3.1, 22.0, -1.0, 19.8, -999, 15.0, 30.0]
print(filtrar_mediciones(datos_sensor))  # Debe retornar [14.5, 18.2, 22.0, 19.8]
````

# Pruebas sugeridas:
# print(calcular_mediana([7, 1, 3, 9, 5]))      # N impar -> Debe retornar 5.0
# print(calcular_mediana([10, 2, 8, 4, 6, 12]))  # N par -> Debe retornar 7.0

---


````python
def calcular_mediana(datos):
    ordenados = sorted(datos)
    n = len(ordenados)

    if n % 2 != 0:
        return float(ordenados[n // 2])
    else:
        m1 = ordenados[(n // 2) - 1]
        m2 = ordenados[n // 2]
        return (m1 + m2) / 2.0



print(calcular_mediana([7, 1, 3, 9, 5]))  
print(calcular_mediana([10, 2, 8, 4, 6, 12])) 
````
