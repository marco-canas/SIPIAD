````python
# Raíz Digital

## 📖 Definición

La **raíz digital** de un número entero no negativo es el valor de un solo dígito que se obtiene al sumar repetidamente los dígitos del número original hasta obtener un único dígito.

En otras palabras: se suman los dígitos del número, y si el resultado tiene más de un dígito, se vuelven a sumar sus dígitos, y así sucesivamente hasta llegar a un solo dígito (del 0 al 9).

---

## 🔢 Ejemplos

### Ejemplo 1: Raíz digital de 9875
```
9875 → 9 + 8 + 7 + 5 = 29
29   → 2 + 9 = 11
11   → 1 + 1 = 2
```
**Raíz digital de 9875 = 2**

### Ejemplo 2: Raíz digital de 456
```
456 → 4 + 5 + 6 = 15
15  → 1 + 5 = 6
```
**Raíz digital de 456 = 6**

### Ejemplo 3: Raíz digital de 7
```
7 → ya es un solo dígito
```
**Raíz digital de 7 = 7**

### Ejemplo 4: Raíz digital de 0
**Raíz digital de 0 = 0** (caso especial)

---

## 🧮 Propiedad matemática importante (Módulo 9)

Existe una relación muy útil con la aritmética modular:

> **La raíz digital de un número `n` es igual a `n mod 9`**, con la única excepción de que si el resultado es 0 y `n > 0`, entonces la raíz digital es 9.

Formalmente:

```
raíz_digital(n) = 1 + ((n - 1) mod 9)   para n > 0
raíz_digital(0) = 0
```

### Verificación con el ejemplo anterior:
- `9875 mod 9 = 9875 - 9·1097 = 9875 - 9873 = 2` ✅ (coincide)
- `456 mod 9 = 456 - 9·50 = 456 - 450 = 6` ✅ (coincide)
- `18 mod 9 = 0`, pero su raíz digital es **9** (por la excepción).

---

## 💻 Implementación en Python (método iterativo)

```python
def raiz_digital(n):
    if n == 0:
        return 0
    while n >= 10:
        suma = 0
        while n > 0:
            suma += n % 10   # toma el último dígito
            n //= 10         # elimina el último dígito
        n = suma
    return n
```

## 💻 Implementación en Python (método con módulo)

```python
def raiz_digital_mod(n):
    if n == 0:
        return 0
    return 1 + (n - 1) % 9
```

---

## 🎯 Aplicaciones

1. **Prueba de divisibilidad entre 9:** si la raíz digital es 9, el número es divisible entre 9.
2. **Prueba del 9 (verificación de operaciones aritméticas):** se usaba antiguamente para detectar errores en sumas y multiplicaciones.
3. **Numerología:** aunque no tiene base matemática seria, es famosa por relacionar la raíz digital con propiedades místicas de los números.
4. **Aritmética modular:** es un ejemplo didáctico perfecto para introducir congruencias.

---

## ⚠️ Observaciones clave

- La raíz digital **siempre está entre 0 y 9**.
- **No es lo mismo que la suma de dígitos**, que puede dar un número mayor a 9.
- Es un proceso **finito**: siempre termina en un solo dígito porque cada iteración reduce drásticamente el tamaño del número.

¿Quieres que te muestre cómo usar la raíz digital dentro de alguno de los 20 ejercicios anteriores (por ejemplo, para optimizar el ejercicio 10)?
````
