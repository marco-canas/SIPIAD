## Bloque 1: Condicionales (`if`, `elif`, `else`)

**Problema 1. Detección de datos faltantes y clasificación de un registro**

- `"Registro utilizable"` si faltan menos del 30 %.
- `"Registro dudoso"` si faltan entre el 30 % y el 60 % (incluido el 30 %).
- `"Registro descartado"` en cualquier otro caso.

**Problema 2. Clasificación de un entero según su tipo**
Dado un entero positivo `n`, determina e imprime si es: *"cuadrado perfecto"* (verifica con `int(n**0.5)**2 == n`), *"potencia de 2"* (verifica con ciclo o con `n & (n-1) == 0`), ambas cosas, o *"ninguna"*. Si `n` no es un entero positivo, imprime un mensaje de error.

**Problema 3. Año bisiesto y calendario de datos**

**Problema 4. Asignación de cuartil**
Dados los tres cuartiles `q1`, `q2`, `q3` de un conjunto de datos y un valor `x`, imprime en qué cuartil cae `x`: *"Cuartil 1"* (x ≤ q1), *"Cuartil 2"* (q1 < x ≤ q2), *"Cuartil 3"* (q2 < x ≤ q3) o *"Cuartil 4"* (x > q3). Valida primero que `q1 ≤ q2 ≤ q3`; si no se cumple, imprime un error.

**Problema 5. Criterios de divisibilidad**

**Problema 12. Sucesión de Fibonacci y razón áurea**
