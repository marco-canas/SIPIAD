## I. Problemas con `if`, `elif` y `else`

### 1. Clasificación de números en un conjunto de datos

Una empresa registra las edades de un grupo de usuarios:

```python
edades = [12, 18, 25, 31, 17, 65, 42, 9, 70, 23]
```

Escriba un programa que recorra la lista y clasifique cada edad en una de las siguientes categorías:

- `"menor de edad"` si es menor que 18.
- `"adulto"` si está entre 18 y 59 años.
- `"adulto mayor"` si es 60 o más.

El programa debe mostrar la edad y su categoría.

**Evalúa:** `for`, `if`, `elif`, `else`.

### 2. Clasificación de números según propiedades aritméticas

```python
datos = [7, 12, 15, 20, 23, 30, 31, 42, 50]
```

```text
30 → par, divisible entre 3 y divisible entre 5
```

```python
cantidad_divisores(n)
```

Por ejemplo:

```python
cantidad_divisores(12)
```

```text
6
```

$$
D(12)=\{1,2,3,4,6,12\}.
$$

Después utilice la función para encontrar, entre los números del 1 al 100, cuál tiene mayor cantidad de divisores.
