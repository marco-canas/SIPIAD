### **Bloque 1: Condicionales (`if`, `elif`, `else`)**

1. **Test de Fermat y Posibles Pseudo-primos (Teoría de Números)**
El pequeño teorema de Fermat establece que si $p$ es primo y $a$ no es divisible por $p$, entonces $a^{p-1} \equiv 1 \pmod p$. Toma valores enteros `a`, `p` y $n = a^{p-1} \pmod p$. Escribe una estructura condicional que verifique:
* Si $n \neq 1$, imprime `"Definitivamente compuesto"`.
* Si $n == 1$ y $p$ es impar, imprime `"Candidato a primo (o pseudoprimo)"`.
* Si $p == 2$, imprime `"Caso base: 2 es primo"`.


2. **Detección de Desbalance de Clases en un Dataset (Ciencia de Datos)**
Dado un diccionario con la cantidad de instancias por clase, por ejemplo `clases = {"clase_A": 850, "clase_B": 120, "clase_C": 30}`, calcula la proporción de la clase minoritaria respecto al total. Usa un condicional para clasificar el dataset:
* **Balanceado**: Proporción minoritaria $\ge 0.30$.
* **Moderadamente Desbalanceado**: $0.10 \le \text{Proporción} < 0.30$.
* **Severamente Desbalanceado**: Proporción $< 0.10$ (sugiere aplicar técnicas como SMOTE o *undersampling*).


3. **Clasificación de Números de Fibonacci por Paridad y Congruencia (Teoría de Números)**
Dado un número `f` que pertenece a la secuencia de Fibonacci, evalúa con condicionales su residuo al dividir entre 4 ($f \pmod 4$):
* Si el residuo es 0, imprime `"Múltiplo de 4 (Posición múltiplo de 6 en Fibonacci)"`.
* Si el residuo es 1 o 3, imprime `"Impar en la secuencia"`.
* Si el residuo es 2, imprime `"Par no múltiplo de 4"`.


4. **Validación de Matriz de Confusión (Ciencia de Datos)**
Dados cuatro valores enteros que representan una matriz de confusión: `TP` (Verdaderos Positivos), `TN` (Verdaderos Negativos), `FP` (Falsos Positivos) y `FN` (Falsos Negativos):
* Si $TP + FN == 0$ o $TN + FP == 0$, imprime `"Error: Divisibilidad por cero en métricas primarias"`.
* De lo contrario, evalúa si la precisión $TP / (TP + FP)$ es mayor al $80\%$; si es así, imprime `"Modelo Aceptable"`, de lo contrario `"Modelo Deficiente"`.


5. **Clasificación de Números Narcisistas / Armstrong (Teoría de Números)**
Dado un entero positivo de tres dígitos `n` (por ejemplo, 153), extrae sus dígitos usando operadores de división entera `//` y módulo `%`. Utiliza un condicional para verificar si la suma de sus cubos es igual al número original ($1^3 + 5^3 + 3^3 = 153$) e imprime si es un **Número Narcisista** o **No Narcisista**.

### **Bloque 2: Bucle `for**`

2. Usa un segundo ciclo `for` para calcular la varianza y la desviación estándar.
3. Usa un tercer ciclo `for` para transformar cada dato a su valor Z-score y guardarlo en una nueva lista.

9. **Codificación de Frecuencias para Variables Categóricas (Ciencia de Datos)**

### **Bloque 3: Bucle `while**`

### **Bloque 4: Funciones de Usuario (`def` y `lambda`)**

15. **Función de Distancia Minkowski Aumentada (Ciencia de Datos)**
Crea una función `def distancia_minkowski(v1, v2, p=2)` que tome dos listas/tuplas numéricas de igual longitud y el parámetro de orden $p$:
* Si $p = 1$, calcula la distancia Manhattan.
* Si $p = 2$, calcula la distancia Euclidiana.
* La función debe retornar la distancia calculada: $D(v1, v2) = \left( \sum \vert{}v1_i - v2_i\vert{}^p \right)^{1/p}$.

* $1$ si $n$ es libre de cuadrados y tiene un número **par** de factores primos distintos.
* $-1$ si $n$ es libre de cuadrados y tiene un número **impar** de factores primos distintos.

Aplica esta función a toda la lista usando `map()` o una comprensión de listas.
