# Recordatorio: Uso de `sorted()` y Definición de Mediana

## ¿Cómo usar `sorted()`?

La función `sorted()` recibe una **lista** (o cualquier iterable) y devuelve una **nueva lista ordenada de menor a mayor** (por defecto). **No modifica la lista original.**

**Sintaxis:**
```python
nueva_lista = sorted(lista_original)
```

**Ejemplo:**
```python
datos = [45, 12, 67, 23, 89, 34]
ordenados = sorted(datos)
print(ordenados)   # [12, 23, 34, 45, 67, 89]
print(datos)       # [45, 12, 67, 23, 89, 34]  (sin cambios)
```

También puede ordenar de **mayor a menor** con el parámetro `reverse=True`:
```python
ordenados_desc = sorted(datos, reverse=True)
print(ordenados_desc)   # [89, 67, 45, 34, 23, 12]
```

> En este parcial **sí está permitido usar `sorted()`** para ordenar datos numéricos.

## ¿Qué es la mediana?

La **mediana** es el valor que ocupa la **posición central** de un conjunto de datos **ordenados de menor a mayor**. Es una medida de tendencia central que **no se ve afectada por valores extremos (outliers)**, a diferencia de la media.

**Reglas para calcularla:**

1. **Ordenar** los datos de menor a mayor (use `sorted()`).
2. Si la cantidad de datos `n` es **impar**:
   
   $$\text{Mediana} = \text{valor en la posición } \frac{n+1}{2}$$

3. Si la cantidad de datos `n` es **par**:
   
   $$\text{Mediana} = \frac{\text{valor en la posición } \frac{n}{2} + \text{valor en la posición } \frac{n}{2}+1}{2}$$

**Ejemplo 1 (n impar):**
```
datos = [45, 12, 67, 23, 89]
ordenados = [12, 23, 45, 67, 89]   # n = 5
Mediana = valor en la posición 3 = 45
```

**Ejemplo 2 (n par):**
```
datos = [45, 12, 67, 23]
ordenados = [12, 23, 45, 67]   # n = 4
Mediana = (23 + 45) / 2 = 34.0
```
