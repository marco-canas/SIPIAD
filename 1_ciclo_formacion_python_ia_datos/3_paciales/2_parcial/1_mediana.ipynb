# El problema de hallar la mediana (la mediana es el dato central que permite establecer cual es el valor que acota superiormente al 50% de los datos)



```python
datos1 = [2, 1, 4, 3]
datos2 =  [2, 1, 4, 3, 5]
```

```python
def calcula_mediana(datos):
    datos_ordenados = sorted(datos)
    num_datos = len(datos)
    if num_datos % 2 == 0: 
        posicion_izq = num_datos//2
        posicion_der = posicion_izq + 1
        indice_izq = posicion_izq - 1
        indice_der = posicion_der - 1
        mediana = (datos_ordenados[indice_izq] + datos_ordenados[indice_der])/2
    else:
        posicion_central = (num_datos+1)//2
        indice_central = posicion_central - 1
        mediana = datos_ordenados[indice_central]
    return datos_ordenados, mediana     
        
```

```python
calcula_mediana(datos2)
```
