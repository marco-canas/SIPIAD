## Bloque 1: Condicionales (`if`, `elif`, `else`)

**Problema 1. Clasificación de un valor atípico (outlier)**
Dados una lista de datos numéricos `datos`, calcula manualmente su media `m` y su desviación estándar `s`. Luego, para un valor `x` ingresado por el usuario, imprime:
- `"Atípico extremo"` si `|x - m| > 3s`
- `"Atípico moderado"` si `2s < |x - m| <= 3s`
- `"Normal"` en cualquier otro caso

**Problema 2. Paridad y divisibilidad**
Dado un entero positivo `n`, imprime:
- `"Divisible por 6"` si es divisible por 2 y por 3.
- `"Par, no divisible por 3"` si es par pero no divisible por 3.
- `"Divisible por 3, impar"` si es divisible por 3 pero impar.
- `"Ninguna de las anteriores"` en otro caso.

**Problema 3. Categorización de un índice de correlación**
Dado un coeficiente de correlación `r` (entre -1 y 1), clasifícalo como: *"Correlación perfecta"* (|r| = 1), *"Fuerte"* (0.7 ≤ |r| < 1), *"Moderada"* (0.4 ≤ |r| < 0.7), *"Débil"* (0 < |r| < 0.4) o *"Nula"* (r = 0). Además, indica si es positiva o negativa. Si `r` está fuera del rango [-1, 1], imprime un mensaje de error.

**Problema 4. Tipo de entero según su resto**
Dado un entero `n`, determina su clasificación según su resto módulo 4 y módulo 5, usando condicionales anidados o encadenados. Imprime, por ejemplo, `"n ≡ 2 (mod 4) y n ≡ 0 (mod 5)"`. Si `n` cumple ambas congruencias `n ≡ 1 (mod 4)` y `n ≡ 1 (mod 5)`, imprime además `"n ≡ 1 (mod 20)"`.

**Problema 5. Validación de un registro de datos**
Un registro es un diccionario con las llaves `"edad"`, `"ingreso"` y `"ciudad"`. Escribe un programa que verifique:
- `edad` es entero entre 0 y 120,
- `ingreso` es numérico y no negativo,
- `ciudad` es una cadena no vacía.

Imprime `"Registro válido"` o un mensaje específico del primer error encontrado. Usa `isinstance`.

## Bloque 2: Ciclos (`for` y `while`)

**Problema 6. Estadísticos descriptivos sin librerías**
Dada una lista de números, usa un ciclo `for` para calcular: la suma, la media, el máximo, el mínimo y la varianza muestral. No uses `sum`, `max`, `min` ni librerías externas.

**Problema 7. Criba de primos**
Usando ciclos `for` y una lista de booleanos, implementa la **Criba de Eratóstenes** para obtener todos los primos menores o iguales a un número `N` dado. Imprime también cuántos primos hay.

**Problema 8. Algoritmo de Euclides**
Con un ciclo `while`, calcula el máximo común divisor de dos enteros `a` y `b` mediante el algoritmo de Euclides. Cuenta además cuántas iteraciones fueron necesarias e imprímelas.

**Problema 9. Conjetura de Collatz**
Dado un entero positivo `n`, usa un ciclo `while` para generar la secuencia de Collatz (si `n` es par, `n/2`; si es impar, `3n+1`) hasta llegar a 1. Imprime la secuencia, su longitud y el valor máximo alcanzado. Luego, con un `for`, encuentra qué número entre 1 y 1000 produce la secuencia más larga.

**Problema 10. Frecuencia de dígitos**
Dado un entero grande `n` (por ejemplo, `2**100`), usa un ciclo para contar cuántas veces aparece cada dígito del 0 al 9 y guárdalo en un diccionario. Imprime el dígito más frecuente y calcula la suma de todos los dígitos de `n` sin convertirlo a cadena (usando `%` y `//`).

**Problema 11. Media móvil**
Dada una serie de tiempo como lista de números y una ventana `k`, genera con ciclos una nueva lista con la media móvil simple. Por ejemplo, para `[2, 4, 6, 8, 10]` y `k = 3`, el resultado es `[4.0, 6.0, 8.0]`. Valida que `k` no sea mayor que la longitud de la lista.

**Problema 12. Números perfectos y amigables**
Un número es *perfecto* si es igual a la suma de sus divisores propios (por ejemplo, 6 y 28). Encuentra con ciclos todos los números perfectos menores que 10 000. Después, encuentra todos los pares de números amigables (a, b) menores que 3000, donde la suma de los divisores propios de a es b y viceversa, con a ≠ b.

**Problema 13. Muestreo hasta convergencia**
Dada la lista `datos`, simula con un ciclo `while` el cálculo acumulado de la media, tomando un elemento a la vez. Detén el ciclo cuando la diferencia entre dos medias acumuladas consecutivas sea menor que `0.01`, o cuando se agoten los datos. Imprime cuántos datos fueron necesarios.

## Bloque 3: Funciones con `def` y `lambda`

**Problema 14. Función `es_primo`**
Define `es_primo(n)` con `def` que retorne `True` si `n` es primo y `False` en caso contrario, probando divisores solo hasta `√n`. Úsala luego para generar la lista de **primos gemelos** (pares `(p, p+2)` ambos primos) menores que 500.

**Problema 15. Normalización de datos**
Define una función `normalizar(datos, metodo="minmax")` que reciba una lista y devuelva una nueva lista normalizada. Si `metodo="minmax"`, aplica `(x - min)/(max - min)`; si `metodo="zscore"`, aplica `(x - media)/desv_est`. Si el método es otro, lanza un mensaje de error. Maneja el caso en que todos los datos sean iguales.

**Problema 16. Factorización prima**
Define `factorizar(n)` que retorne un diccionario con la factorización prima de `n`, donde las llaves son los primos y los valores sus exponentes. Por ejemplo, `factorizar(360)` → `{2: 3, 3: 2, 5: 1}`. Con base en ella, define `num_divisores(n)` que calcule cuántos divisores positivos tiene `n` usando la fórmula del producto de `(exponente + 1)`.

**Problema 17. Funciones `lambda` con `map`, `filter` y `sorted`**
Dada la lista `datos = [12, 7, 3, 18, 25, 9, 14, 31, 6]`, resuelve usando **únicamente** expresiones `lambda`:
- a) Obtén los cuadrados de todos los elementos (`map`).
- b) Filtra los números que son múltiplos de 3 (`filter`).
- c) Ordena la lista según el residuo de dividir cada elemento entre 5 (`sorted` con `key`).
- d) Dada una lista de tuplas `(nombre, calificacion)`, ordénala de mayor a menor calificación.

**Problema 18. Inverso modular y congruencias**
Define `inverso_modular(a, m)` que retorne el inverso multiplicativo de `a` módulo `m` (es decir, `x` tal que `a·x ≡ 1 (mod m)`), o `None` si no existe (cuando `mcd(a, m) ≠ 1`). Usa una función auxiliar `mcd` y un ciclo. Luego resuelve la congruencia `7x ≡ 3 (mod 11)`.

**Problema 19. Fábrica de funciones y regresión lineal simple**
a) Define `crear_modelo(pendiente, intercepto)` que retorne una función `lambda` que calcule `y = pendiente·x + intercepto`.
b) Define `ajustar_recta(x, y)` que calcule por mínimos cuadrados la pendiente y el intercepto a partir de dos listas, y retorne el modelo creado con `crear_modelo`.
c) Define `error_cuadratico_medio(modelo, x, y)` y evalúa el ajuste con los datos `x = [1, 2, 3, 4, 5]`, `y = [2.1, 3.9, 6.2, 7.8, 10.1]`.

**Problema 20. Función de Euler y análisis de datos numéricos**
Define `phi(n)` que calcule la función φ de Euler (cantidad de enteros entre 1 y n coprimos con n). Luego:
- a) Calcula `phi(n)` para `n` de 1 a 50 y guarda los resultados en un diccionario.
- b) Con `lambda`, encuentra el `n` en ese rango con mayor razón `phi(n)/n` y el que tiene la menor.
- c) Verifica con un ciclo la identidad `sum(phi(d) for d in divisores de n) == n` para todo `n` entre 1 y 50, e imprime `"Identidad verificada"` si se cumple siempre.
