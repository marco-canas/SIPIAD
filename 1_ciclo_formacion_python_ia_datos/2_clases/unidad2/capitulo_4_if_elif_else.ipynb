## 🔍 1. Conceptos Básicos

### ¿Qué es una condición?

Una **condición** es una expresión que se evalúa como `True` (verdadero) o `False` (falso).

```python
edad = 25
condicion = edad >= 18  # True
```

### Tabla de Verdad

| Operación | Símbolo | Ejemplo | Resultado |
|-----------|---------|---------|-----------|
| Igualdad | `==` | `5 == 5` | `True` |
| Desigualdad | `!=` | `5 != 3` | `True` |
| Mayor que | `>` | `5 > 3` | `True` |
| Menor que | `<` | `5 < 3` | `False` |
| Mayor o igual | `>=` | `5 >= 5` | `True` |
| Menor o igual | `<=` | `5 <= 3` | `False` |

````python
# 🧪 Ejemplo interactivo
print("=== Evaluación de Condiciones ===\n")

x = 10
y = 20

condiciones = [
    (x == 10, f"{x} == 10"),
    (x != y, f"{x} != {y}"),
    (x > 5, f"{x} > 5"),
    (y < 15, f"{y} < 15"),
    (x >= 10, f"{x} >= 10"),
    (y <= 20, f"{y} <= 20")
]

for resultado, descripcion in condiciones:
    print(f"✅ {descripcion} → {resultado}")
````

# Visualización de múltiples condiciones

````python
# Función por partes
def f_por_partes(x):
    if x < -2:
        return -2
    elif -2 <= x < 0:
        return x
    elif 0 <= x < 2:
        return x**2
    else:
        return 2

# Visualización
x_vals = np.linspace(-4, 4, 1000)
y_vals = [f_por_partes(x) for x in x_vals]

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', linewidth=2)
plt.axhline(y=0, color='black', linewidth=0.5)
plt.axvline(x=0, color='black', linewidth=0.5)

# Regiones
plt.axvspan(-4, -2, alpha=0.2, color='red', label='x < -2: -2')
plt.axvspan(-2, 0, alpha=0.2, color='green', label='-2 ≤ x < 0: x')
plt.axvspan(0, 2, alpha=0.2, color='blue', label='0 ≤ x < 2: x²')
plt.axvspan(2, 4, alpha=0.2, color='purple', label='x ≥ 2: 2')

plt.title('Función Definida por Partes con elif')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

### 6.1 Clasificación con Umbral (Regresión Logística Simulada)

````python
# Simulación de predicciones de un modelo
np.random.seed(42)
n_muestras = 200
predicciones = np.random.rand(n_muestras)

def clasificar_umbral(probabilidad, umbral=0.5):
    if probabilidad >= umbral:
        return "Clase 1 (Positivo)"
    else:
        return "Clase 0 (Negativo)"

# Aplicar clasificación
clases = [clasificar_umbral(p) for p in predicciones]

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Histograma de probabilidades
ax1.hist(predicciones, bins=20, edgecolor='black', alpha=0.7)
ax1.axvline(x=0.5, color='red', linestyle='--', linewidth=2, label='Umbral=0.5')
ax1.set_title('Distribución de Probabilidades')
ax1.set_xlabel('Probabilidad')
ax1.set_ylabel('Frecuencia')
ax1.legend()

# Scatter de resultados
colores = ['blue' if c == 'Clase 0 (Negativo)' else 'orange' for c in clases]
ax2.scatter(range(n_muestras), predicciones, c=colores, alpha=0.6, s=50)
ax2.axhline(y=0.5, color='red', linestyle='--', linewidth=2, label='Umbral=0.5')
ax2.set_title('Clasificación Binaria')
ax2.set_xlabel('Muestra')
ax2.set_ylabel('Probabilidad')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Estadísticas
n_clase1 = sum(1 for c in clases if c == 'Clase 1 (Positivo)')
n_clase0 = n_muestras - n_clase1
print(f"📊 Resultados de clasificación:")
print(f"  • Clase 1 (Positivo): {n_clase1} muestras ({n_clase1/n_muestras*100:.1f}%)")
print(f"  • Clase 0 (Negativo): {n_clase0} muestras ({n_clase0/n_muestras*100:.1f}%)")
````

### 6.2 Segmentación de Clientes (RFM Simulado)

````python
# Datos simulados de clientes
import pandas as pd 
clientes = pd.DataFrame({
    'Cliente': [f'Cliente_{i}' for i in range(1, 21)],
    'Ingreso': np.random.randint(20, 150, 20) * 1000,
    'Edad': np.random.randint(18, 70, 20)
})
clientes 
````

````python

def segmentar_cliente(ingreso, edad):
    if ingreso > 100_000 and edad > 40:
        return '🎯 VIP Premium'
    elif ingreso > 50_000:
        return '💰 Clase Media Alta'
    elif edad < 30:
        return '🧑‍🎓 Joven Profesional'
    elif ingreso < 30_000:
        return '📊 Clase Media Baja'
    else:
        return '🏠 Clase Media'
````

````python

# Aplicar segmentación
clientes['Segmento'] = clientes.apply(
    lambda row: segmentar_cliente(row['Ingreso'], row['Edad']), axis=1
)

# Mostrar resultados
print("=== SEGMENTACIÓN DE CLIENTES ===\n")
display(clientes)

# Estadísticas por segmento
print("\n=== ESTADÍSTICAS POR SEGMENTO ===")
segmentos = clientes['Segmento'].value_counts()
for segmento, count in segmentos.items():
    print(f"  • {segmento}: {count} clientes ({count/len(clientes)*100:.1f}%)")
````

### 6.3 Detección de Anomalías (Z-score)

````python
# Datos de temperatura
import numpy as np 
np.random.seed(123)
temperaturas = np.random.normal(25, 5, 1000)  # Media 25°C, desviación 5°C
temperaturas = np.append(temperaturas, [80, 85, 90])  # Añadir anomalías

def detectar_anomalia(valor, media, desviacion, umbral=2.5):
    z_score = abs((valor - media) / desviacion)
    if z_score > umbral:
        return '⚠️ Anomalía'
    else:
        return '✅ Normal'

# Calcular estadísticas
media = np.mean(temperaturas)
desviacion = np.std(temperaturas)

# Detectar anomalías
resultados = [detectar_anomalia(t, media, desviacion) for t in temperaturas]
resultados 
````

````python

# Visualización
plt.figure(figsize=(12, 6))
colores = ['red' if r == '⚠️ Anomalía' else 'blue' for r in resultados]
plt.scatter(range(len(temperaturas)), temperaturas, c=colores, alpha=0.6, s=20)
plt.axhline(y=media, color='green', linestyle='-', linewidth=2, label='Media')
plt.axhline(y=media + 2.5*desviacion, color='orange', linestyle='--', label='Umbral')
plt.axhline(y=media - 2.5*desviacion, color='orange', linestyle='--')

plt.title('Detección de Anomalías en Temperatura')
plt.xlabel('Muestra')
plt.ylabel('Temperatura (°C)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

````python

# Estadísticas
n_anomalias = sum(1 for r in resultados if r == '⚠️ Anomalía')
print(f"📊 Resultados de detección:")
print(f"  • Media: {media:.2f}°C")
print(f"  • Desviación estándar: {desviacion:.2f}°C")
print(f"  • Anomalías detectadas: {n_anomalias}")
````

### Ejercicio 1: Calculadora de IMC (Índice de Masa Corporal)

````python
# Calculadora IMC
def calcular_imc(peso, altura):
    imc = peso / (altura ** 2)
    
    if imc < 18.5:
        categoria = "Bajo peso"
        recomendacion = "🍎 Necesitas ganar peso saludablemente"
    elif imc < 25:
        categoria = "Peso normal"
        recomendacion = "🌟 ¡Excelente! Mantén tu estilo de vida"
    elif imc < 30:
        categoria = "Sobrepeso"
        recomendacion = "🏃‍♂️ Aumenta actividad física y revisa dieta"
    elif imc < 35:
        categoria = "Obesidad I"
        recomendacion = "⚠️ Consulta a un especialista en nutrición"
    elif imc < 40:
        categoria = "Obesidad II"
        recomendacion = "🚨 Necesitas atención médica profesional"
    else:
        categoria = "Obesidad III"
        recomendacion = "🆘 Busca ayuda médica inmediata"
    
    return imc, categoria, recomendacion

# Prueba
peso = float(input("Ingresa tu peso (kg): "))
altura = float(input("Ingresa tu altura (m): "))

imc, categoria, recomendacion = calcular_imc(peso, altura)

print("\n=== RESULTADOS IMC ===")
print(f"📊 IMC: {imc:.1f}")
print(f"📋 Categoría: {categoria}")
print(f"💡 Recomendación: {recomendacion}")
````

### Ejercicio 2: Sistema de Préstamos

````python
# Sistema de aprobación de préstamos
def evaluar_prestamo(ingreso, score_credito, antiguedad_laboral):
    aprobado = True
    razones = []
    
    if ingreso < 20000:
        aprobado = False
        razones.append(f"Ingreso insuficiente (${ingreso:,} - mínimo $20,000)")
    
    if score_credito < 700:
        aprobado = False
        razones.append(f"Score crediticio bajo ({score_credito} - mínimo 700)")
    
    if antiguedad_laboral < 1:
        aprobado = False
        razones.append(f"Antigüedad laboral insuficiente ({antiguedad_laboral} años - mínimo 1 año)")
    
    # Si pasa todos los filtros, verificar condiciones especiales
    if aprobado:
        if ingreso >= 50000 and score_credito >= 750:
            categoria = "🏆 Cliente Premium - Aprobado con ventajas"
        elif ingreso >= 35000:
            categoria = "✅ Aprobado - Tasa estándar"
        else:
            categoria = "✅ Aprobado - Tasa preferencial"
    else:
        categoria = "❌ Rechazado"
    
    return aprobado, categoria, razones

# Prueba
ingreso = float(input("Ingreso mensual ($): "))
score = int(input("Score crediticio (300-850): "))
antiguedad = float(input("Años en trabajo actual: "))

aprobado, categoria, razones = evaluar_prestamo(ingreso, score, antiguedad)

print("\n=== EVALUACIÓN DE PRÉSTAMO ===")
print(f"📊 Resultado: {categoria}")
if not aprobado:
    print("\n📋 Motivos del rechazo:")
    for i, razon in enumerate(razones, 1):
        print(f"  {i}. {razon}")
````

## 🧪 8. Proyecto Integrador: Sistema de Recomendación de Productos

### Contexto

Crea un sistema que recomiende productos a clientes basado en sus características:

````python
# Sistema de recomendación de productos
def recomendar_producto(edad, genero, interes, presupuesto):
    recomendaciones = []
    
    # Recomendaciones basadas en edad
    if edad < 18:
        recomendaciones.append("🎮 Videojuegos educativos")
        if presupuesto < 100:
            recomendaciones.append("📚 Libros de aventura")
    elif edad < 30:
        recomendaciones.append("📱 Tecnología y gadgets")
        if interes == "deportes":
            recomendaciones.append("🏋️ Equipo de fitness")
        elif interes == "moda":
            recomendaciones.append("👕 Ropa de moda urbana")
    elif edad < 50:
        recomendaciones.append("🏠 Productos para el hogar")
        if presupuesto > 500:
            recomendaciones.append("🛋️ Muebles de diseño")
        else:
            recomendaciones.append("🧹 Accesorios para el hogar")
    else:
        recomendaciones.append("🌿 Productos de bienestar")
        if interes == "salud":
            recomendaciones.append("💊 Suplementos nutricionales")
        else:
            recomendaciones.append("📖 Libros de lectura ligera")
    
    # Recomendaciones basadas en género
    if genero.lower() == "femenino":
        recomendaciones.append("💄 Cosméticos y cuidado personal")
    elif genero.lower() == "masculino":
        recomendaciones.append("🔧 Productos de cuidado personal")
    
    # Filtro por presupuesto
    if presupuesto < 50:
        recomendaciones = [r for r in recomendaciones if not any(
            palabra in r for palabra in ['Muebles', 'Tecnología', 'gadgets', 'diseño']
        )]
        recomendaciones.append("🛒 Productos básicos")
    
    # Limitar a 5 recomendaciones únicas
    return list(dict.fromkeys(recomendaciones))[:5]

# Interfaz de usuario
print("=== SISTEMA DE RECOMENDACIÓN DE PRODUCTOS ===\n")
print("Responde las siguientes preguntas para obtener recomendaciones personalizadas:\n")

edad = int(input("Ingresa tu edad: "))
genero = input("Ingresa tu género (Masculino/Femenino/Otro): ")
interes = input("¿Cuál es tu principal interés? (deportes/moda/tecnología/salud/otro): ")
presupuesto = float(input("¿Cuánto planeas gastar? ($): "))

# Generar recomendaciones
recomendaciones = recomendar_producto(edad, genero, interes, presupuesto)

# Mostrar resultados
print("\n" + "="*50)
print("🌟 TUS RECOMENDACIONES PERSONALIZADAS 🌟")
print("="*50)
for i, recomendacion in enumerate(recomendaciones, 1):
    print(f"{i}. {recomendacion}")

print("\n💡 Consejos:")
if presupuesto < 50:
    print("  • Busca ofertas y descuentos en productos básicos")
elif presupuesto < 200:
    print("  • Considera productos en oferta o paquetes económicos")
else:
    print("  • Puedes considerar productos premium o ediciones especiales")
````

**Enunciado:** Implementa una función que clasifique una fruta según su color y temporada.

````python
# Tu solución aquí
def clasificar_fruta(color, temporada):
    if color == "rojo" and temporada == "verano":
        return "🍓 Fresa - Excelente para jugos"
    elif color == "amarillo" and temporada == "verano":
        return "🍌 Plátano - Rico en potasio"
    elif color == "naranja" and temporada == "invierno":
        return "Naranja - Fuente de vitamina C"
    elif color == "verde" and temporada == "primavera":
        return "Manzana verde - Baja en azúcar"
    else:
        return "Fruta no reconocida o fuera de temporada"

# Prueba
print(clasificar_fruta("rojo", "verano"))
print(clasificar_fruta("amarillo", "invierno"))
````
