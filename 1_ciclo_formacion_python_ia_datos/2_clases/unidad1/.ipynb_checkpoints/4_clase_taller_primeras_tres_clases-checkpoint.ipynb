
### Módulo 3 — Sets y diccionarios


````python
# M3_sets_diccionarios.ipynb

# --- SETS (conjuntos: únicos, no ordenados) ---
municipios_antioquia = {"Caucasia", "El Bagre", "Medellín", "Nechí", "Zaragoza"}
municipios_cordoba = {"Montería", "Cereté", "Caucasia", "Sahagún", 'la Apartada'}

print("Antioquia:", municipios_antioquia)
print("Córdoba:", municipios_cordoba)
````

````python

# Operaciones de conjuntos
print("Intersección:", municipios_antioquia & municipios_cordoba)
print("Unión:", municipios_antioquia | municipios_cordoba)
print("Diferencia (A - B):", municipios_antioquia - municipios_cordoba)
print("Diferencia simétrica:", municipios_antioquia ^ municipios_cordoba)
````

````python

# Agregar y eliminar
municipios_antioquia.add("Tarazá")
municipios_antioquia.discard("Medellín")
print("Actualizado Municipios Antioquia:", municipios_antioquia)
````

````python

# --- DICCIONARIOS (clave-valor) ---
poblacion_municipios = {
    "Caucasia": 122_000,
    "El Bagre": 60_000,
    "Zaragoza": 32_000,
    "Nechí": 30_000,
    "Tarazá": 50_000
}

print("Población de Caucasia:", poblacion_municipios["Caucasia"])
````

````python

# Métodos útiles
print("Claves:", list(poblacion_municipios.keys()))
````

````python
print("Valores:", list(poblacion_municipios.values()))
````

````python
print("Ítems:", list(poblacion_municipios.items()))
````

````python

# Agregar / modificar
poblacion_municipios["Cáceres"] = 40_000 # agregar
poblacion_municipios["Zaragoza"] = 33_000  # actualización
print("Diccionario actualizado:", poblacion_municipios)
````

````python

# Recorrido
for municipio, poblacion in poblacion_municipios.items():
    print(f"{municipio}: {poblacion:,} habitantes")
````


### Módulo 4 — Transición a `nd-array` y `DataFrame`


````python
# M4_numpy_pandas.ipynb
import numpy as np
import pandas as pd
````

````python

# --- ND-ARRAY (NumPy) ---
# Creación de un array unidimensional
poblaciones = np.array([122000, 60000, 32000, 30000, 50000])
print("Array 1D:", poblaciones)
print("Tipo:", type(poblaciones))
print("Dimensiones:", poblaciones.ndim)
print("Forma:", poblaciones.shape)
````

````python

# Operaciones vectorizadas (ventaja clave de NumPy)
print("Población total:", poblaciones.sum())
print("Promedio:", poblaciones.mean())
print("Máximo:", poblaciones.max())
print("Mínimo:", poblaciones.min())
print("Poblaciones en miles:", poblaciones / 1000)
````

````python

# Array 2D
matriz_datos = np.array([
    [122000, 2023],
    [60000, 2023],
    [32000, 2023],
    [30000, 2023],
    [50000, 2023]
])
print("Array 2D:\n", matriz_datos)
print("Dimensiones:", matriz_datos.ndim)
print("Forma:", matriz_datos.shape)
print("Número de elementos:", matriz_datos.size)
print("Tipo de datos:", matriz_datos.dtype)
print("Tamaño de cada elemento (bytes):", matriz_datos.itemsize)
print("Tamaño total (bytes):", matriz_datos.nbytes)
print("Transpuesta:\n", matriz_datos.T)
print("Suma por columnas:", matriz_datos.sum(axis=0))
print("Suma por filas:", matriz_datos.sum(axis=1))
print("Forma:", matriz_datos.shape)
````

````python

# --- DATAFRAME (Pandas) ---
datos = {
    "municipio": ["Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá"],
    "poblacion": [122000, 60000, 32000, 30000, 50000],
    "departamento": ["Antioquia", "Antioquia", "Antioquia", "Antioquia", "Antioquia"]
}
df = pd.DataFrame(datos)
print("\nDataFrame:\n", df)
````

## Problema 1 — "Ficha de un municipio del Bajo Cauca" (Tipos de datos y diccionarios)
[Video de apoyo]()
**Contexto:** El semillero trabajará con datos del territorio. Necesitas crear una ficha básica de un municipio.

**Enunciado:**
1. Crea variables con los siguientes tipos de datos para el municipio de **Caucasia**:
   - `nombre` (str)
   - `poblacion` (int)
   - `area_km2` (float)
   - `es_capital_bajo_cauca` (bool)
2. Crea un **diccionario** llamado `ficha_caucasia` que almacene estas variables con claves descriptivas.
3. Calcula e imprime la **densidad poblacional** (`poblacion / area_km2`) usando los valores del diccionario.
4. Imprime la ficha completa de forma legible (usa `f-strings`).
5. Usa un **operador lógico** para imprimir si Caucasia es capital del Bajo Cauca **y** tiene más de 100,000 habitantes.

**Pista:** Usa `ficha_caucasia["poblacion"]` para acceder a los valores.

---


````python
# 1. Variables con sus tipos de datos
nombre = "Caucasia"                    # str
poblacion = 120_000                     # int (aprox.)
area_km2 = 1411.0                      # float (km²)
es_capital_bajo_cauca = True           # bool

# 2. Diccionario con claves descriptivas
ficha_caucasia = {
    "nombre": nombre,
    "poblacion": poblacion,
    "area_km2": area_km2,
    "es_capital_bajo_cauca": es_capital_bajo_cauca,
}

# 3. Densidad poblacional usando los valores del diccionario
densidad = ficha_caucasia["poblacion"] / ficha_caucasia["area_km2"]
print(f"Densidad poblacional: {densidad:.2f} hab/km²")
````

````python

# 4. Ficha completa de forma legible con f-strings
print("\n--- FICHA DEL MUNICIPIO ---")
print(f"Nombre:            {ficha_caucasia['nombre']}")
print(f"Población:         {ficha_caucasia['poblacion']:,} habitantes")
print(f"Área:              {ficha_caucasia['area_km2']} km²")
print(f"¿Capital del Bajo Cauca?: {ficha_caucasia['es_capital_bajo_cauca']}")
````

````python

# 5. Operador lógico: es capital Y tiene más de 100,000 habitantes
es_capital_y_grande = (
    ficha_caucasia["es_capital_bajo_cauca"]
    and ficha_caucasia["poblacion"] > 100_000
)
print(f"\n¿Es capital del Bajo Cauca y tiene más de 100,000 habitantes? {es_capital_y_grande}")
````


## Problema 2 — "Lista de municipios y operaciones" (Listas, slicing y métodos)
[Video de apoyo al aprendizaje](https://www.youtube.com/watch?v=NGl90OnXAmI)


**Contexto:** Se requiere organizar una lista de municipios del Bajo Cauca antioqueño con sus poblaciones proyectadas por el DANE para el año 2026, con el fin de realizar un análisis preliminar.

**Enunciado:**
1. Crea dos **listas** paralelas:
   - `municipios = ["Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá"]`
   - `poblaciones = [93843, 56197, 26996, 29853, 29542]`
2. Imprime el **primer** y el **último** municipio usando indexación negativa.
3. Usa **slicing** para obtener los **tres primeros** municipios y los **dos últimos**.
4. Agrega el municipio `"Cáceres"` con población `28944` a ambas listas.
5. Elimina `"Nechí"` y su población correspondiente de ambas listas (¡mantén la correspondencia!).
6. Calcula e imprime la **población total** y el **promedio** usando `sum()` y `len()`.
7. Imprime cuántos municipios tienen población **mayor a 50,000** (usa un ciclo `for` o una comprensión de listas).

**Reto extra:** Ordena los municipios por población de mayor a menor usando `sorted()` con `zip`.

---



### Nota sobre las fuentes de datos

Las cifras de población utilizadas corresponden a proyecciones del DANE para 2025-2026, publicadas por fuentes que citan dicha entidad oficial:

| Municipio | Población proyectada | Año de referencia | Fuente |
|-----------|----------------------|-------------------|--------|
| Caucasia | 93.843 | 2025 | DANE (vía Telencuestas)  |
| El Bagre | 56.197 | 2025 | DANE (vía Telencuestas)  |
| Zaragoza | 26.996 | 2026 | DANE (vía Telencuestas)  |
| Nechí | 29.853 | 2026 | DANE (vía Telencuestas)  |
| Tarazá | 29.542 | 2026 | DANE (vía Telencuestas)  |
| Cáceres | 28.944 | 2026 | DANE (vía Telencuestas)  |



> **Observación importante:** Las cifras originales del enunciado (122000, 60000, 32000, 30000, 50000) no coinciden con las proyecciones oficiales del DANE. Las cifras actualizadas reflejan datos más cercanos a las estimaciones oficiales. Si se desea un año de referencia uniforme, se pueden ajustar todas las poblaciones al año 2026 (las cifras de Caucasia y El Bagre disponibles son de 2025; sus proyecciones para 2026 serían ligeramente distintas, pero no se encontraron en los resultados de búsqueda de DeepSeek).


````python
# 1. Listas paralelas con datos DANE (proyecciones 2025-2026)
municipios = ["Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá"]
poblaciones = [93843, 56197, 26996, 29853, 29542]

print("Listas iniciales:")
print("Municipios: ", municipios)
print("Poblaciones:", poblaciones)
````

````python

# 2. Primer y último municipio con indexación negativa
print("\n--- Indexación negativa ---")
print(f"Primer municipio:  {municipios[0]}  (o con negativo: {municipios[-5]})")
print(f"Último municipio:  {municipios[-1]}")
````

````python

# 3. Slicing: tres primeros y dos últimos
print("\n--- Slicing ---")
print(f"Tres primeros municipios: {municipios[:3]}")
print(f"Dos últimos municipios:   {municipios[-2:]}")
````

````python

# 4. Agregar "Cáceres" con población 28944 a ambas listas
municipios.append("Cáceres")
poblaciones.append(28944)
print("\n--- Después de agregar Cáceres ---")
print("Municipios: ", municipios)
print("Poblaciones:", poblaciones)
````

````python

# 5. Eliminar "Nechí" y su población (mantener correspondencia)
indice_nechi = municipios.index("Nechí")
indice_nechi  
````

````python
municipios.pop(indice_nechi)
poblaciones.pop(indice_nechi)
````

````python
print("\n--- Después de eliminar Nechí ---")
print("Municipios: ", municipios)
print("Poblaciones:", poblaciones)
````

````python

# 6. Población total y promedio con sum() y len()
total = sum(poblaciones)
promedio = total / len(poblaciones)
print("\n--- Estadísticas ---")
print(f"Población total:   {total:,} habitantes")
print(f"Población promedio: {promedio:,.2f} habitantes")
````

````python
list(zip(municipios, poblaciones)) # crea un iterador de tuplas (municipio, población)
````

````python

# 7. Cuántos municipios tienen población mayor a 50,000
mayores_50k = [m for m, p in zip(municipios, poblaciones) if p > 50_000]
print("\n--- Municipios con más de 50,000 habitantes ---")
print(f"Cantidad: {len(mayores_50k)}")
print(f"Municipios: {mayores_50k}")
````

````python

# Reto extra: ordenar por población de mayor a menor usando sorted() con zip
print("\n--- Reto extra: ranking por población (mayor a menor) ---")
ranking = sorted(zip(municipios, poblaciones), key=lambda x: x[1], reverse=True)
for posicion, (municipio, poblacion) in enumerate(ranking, start=1):
    print(f"{posicion}. {municipio:<12} {poblacion:>7,} habitantes")
````

````python
"""
Gráficos estadísticos - Municipios del Bajo Cauca (datos DANE 2025-2026)
Requiere: matplotlib  (pip install matplotlib)
"""

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker # este paquete sirve para formatear los ejes 
# de los gráficos

# ============================================================
# DATOS (mismos del Problema 2, después de agregar Cáceres y quitar Nechí)
# ============================================================
municipios = ["Caucasia", "El Bagre", "Zaragoza", "Tarazá", "Cáceres"]
poblaciones = [93843, 56197, 26996, 29542, 28944]

# Cálculos auxiliares
total = sum(poblaciones)
promedio = total / len(poblaciones)
umbral = 50000

# Ordenar de mayor a menor para el ranking
ranking = sorted(zip(municipios, poblaciones), key=lambda x: x[1], reverse=True)
nombres_ord = [m for m, p in ranking]
pob_ord = [p for m, p in ranking]

# Paleta de colores (verde para los que superan el umbral, gris para el resto)
colores = ["#2E7D32" if p > umbral else "#9E9E9E" for p in pob_ord]

# ============================================================
# FIGURA 1: Ranking de población (barras horizontales)
# ============================================================
fig1, ax1 = plt.subplots(figsize=(6, 3))
barras = ax1.barh(nombres_ord[::-1], pob_ord[::-1], color=colores[::-1], edgecolor="black")
ax1.set_title("Población por municipio (DANE 2025-2026)", fontsize=14, fontweight="bold")
ax1.set_xlabel("Habitantes")
ax1.xaxis.set_major_formatter(mticker.StrMethodFormatter("{x:,.0f}"))
ax1.axvline(umbral, color="red", linestyle="--", linewidth=1.5, label=f"Umbral {umbral:,}")
ax1.axvline(promedio, color="blue", linestyle=":", linewidth=1.5, label=f"Promedio {promedio:,.0f}")

# Etiquetas de valor al final de cada barra
for barra in barras:
    ancho = barra.get_width()
    ax1.text(ancho + 1500, barra.get_y() + barra.get_height()/2,
             f"{ancho:,.0f}", va="center", fontsize=9)

ax1.legend(loc="lower right")
ax1.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig("grafico1_ranking.png", dpi=150)
plt.show()
````

````python

# ============================================================
# FIGURA 2: Comparación con el promedio (barras verticales)
# ============================================================
fig2, ax2 = plt.subplots(figsize=(4, 3))
colores_prom = ["#2E7D32" if p > promedio else "#C62828" for p in poblaciones]
barras2 = ax2.bar(municipios, poblaciones, color=colores_prom, edgecolor="black")
ax2.axhline(promedio, color="blue", linestyle="--", linewidth=1.8,
            label=f"Promedio = {promedio:,.0f}")
ax2.set_title("Población vs. promedio regional", fontsize=14, fontweight="bold")
ax2.set_ylabel("Habitantes")
ax2.yaxis.set_major_formatter(mticker.StrMethodFormatter("{x:,.0f}"))
ax2.set_ylim(0, max(poblaciones) * 1.15)

for barra in barras2:
    alto = barra.get_height()
    ax2.text(barra.get_x() + barra.get_width()/2, alto + 1200,
             f"{alto:,.0f}", ha="center", fontsize=9)

ax2.legend()
ax2.grid(axis="y", alpha=0.3)
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("grafico2_promedio.png", dpi=150)
plt.show()
````

````python

# ============================================================
# FIGURA 3: Distribución porcentual (torta)
# ============================================================
fig3, ax3 = plt.subplots(figsize=(4, 4))
colores_torta = plt.cm.Set3.colors
explode = [0.08 if p == max(poblaciones) else 0 for p in poblaciones]

def autopct_format(pct):
    return f"{pct:.1f}%"

ax3.pie(poblaciones, labels=municipios, autopct=autopct_format, startangle=90,
        colors=colores_torta, explode=explode,
        wedgeprops={"edgecolor": "black", "linewidth": 0.8})
ax3.set_title(f"Participación porcentual en la población total\n(Total = {total:,} hab)",
              fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("grafico3_torta.png", dpi=150)
plt.show()
````

````python

# ============================================================
# FIGURA 4: Cumplimiento del umbral de 50,000 habitantes
# ============================================================
fig4, ax4 = plt.subplots(figsize=(4, 3))
supera = sum(1 for p in poblaciones if p > umbral)
no_supera = len(poblaciones) - supera

categorias = [f"Supera {umbral:,}", f"No supera {umbral:,}"]
valores = [supera, no_supera]
colores_umbral = ["#2E7D32", "#9E9E9E"]

barras4 = ax4.bar(categorias, valores, color=colores_umbral, edgecolor="black", width=0.5)
ax4.set_title(f"Municipios según umbral de {umbral:,} habitantes", fontsize=14, fontweight="bold")
ax4.set_ylabel("Cantidad de municipios")
ax4.set_ylim(0, max(valores) + 1.5)

for barra in barras4:
    alto = barra.get_height()
    ax4.text(barra.get_x() + barra.get_width()/2, alto + 0.1,
             f"{int(alto)}", ha="center", fontsize=12, fontweight="bold")

ax4.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("grafico4_umbral.png", dpi=150)
plt.show()
````

## Problema 3 — "Conjuntos de municipios por departamento" (Sets y operaciones de conjuntos)
[Video de apoyo al aprendizaje de la solución de este problema]()

**Contexto:** El Bajo Cauca y el Sur de Córdoba comparten dinámicas territoriales. Necesitas identificar municipios comunes y diferencias.

**Enunciado:**
1. Crea dos **sets**:
   - `municipios_antioquia = {"Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá", "Cáceres"}`
   - `municipios_cordoba = {"Montería", "Cereté", "Caucasia", "Sahagún", "Montelíbano"}`
2. Imprime la **intersección** (municipios en ambos departamentos).
3. Imprime la **unión** (todos los municipios sin repetir).
4. Imprime los municipios **exclusivos de Antioquia** y los **exclusivos de Córdoba**.
5. Agrega `"San José de Uré"` a `municipios_cordoba` y elimina `"Sahagún"`.
6. Imprime cuántos municipios hay en total en cada set después de los cambios.
7. **Reflexión escrita (3 líneas):** ¿Por qué un `set` es útil cuando se quieren evitar duplicados? Da un ejemplo aplicado a datos del territorio.

**Pista:** Usa `&`, `|`, `-`, `^`, `.add()`, `.discard()`.

# Solución — Problema 3: Conjuntos de municipios por departamento



````python
# 1. Crear los dos sets
municipios_antioquia = {"Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá", "Cáceres"}
municipios_cordoba = {"Montería", "Cereté", "Caucasia", "Sahagún", "Montelíbano"}
````

````python
print("Antioquia:", municipios_antioquia)
print("Córdoba:  ", municipios_cordoba)
````

````python
# 2. Intersección (municipios en ambos departamentos)
interseccion = municipios_antioquia & municipios_cordoba
print("\n--- Intersección (en ambos departamentos) ---")
print(interseccion)
````

````python
# 3. Unión (todos los municipios sin repetir)
union = municipios_antioquia | municipios_cordoba
print("\n--- Unión (todos sin repetir) ---")
print(union)
print(f"Total de municipios únicos: {len(union)}")
````

````python
# 4. Exclusivos de cada departamento
solo_antioquia = municipios_antioquia - municipios_cordoba
solo_cordoba = municipios_cordoba - municipios_antioquia
print("\n--- Exclusivos de Antioquia ---")
print(solo_antioquia)
print("\n--- Exclusivos de Córdoba ---")
print(solo_cordoba)
````

````python
# (Opcional) Diferencia simétrica: exclusivos de uno u otro, pero no ambos
diferencia_simetrica = municipios_antioquia ^ municipios_cordoba
print("\n--- Diferencia simétrica (^) ---")
print(diferencia_simetrica)
````

````python
# 5. Agregar "San José de Uré" y eliminar "Sahagún" de Córdoba
municipios_cordoba.add("San José de Uré")
municipios_cordoba.discard("Sahagún")
print("\n--- Córdoba después de los cambios ---")
print(municipios_cordoba)
````

````python
# 6. Cantidad de municipios en cada set después de los cambios
print("\n--- Conteo final ---")
print(f"Antioquia: {len(municipios_antioquia)} municipios")
print(f"Córdoba:   {len(municipios_cordoba)} municipios")
````


> **Nota:** El orden de los elementos dentro de un `set` no es determinista (no garantiza orden de inserción), por lo que la salida puede variar en el orden de los nombres, pero **no en el contenido**.

## Puntos clave de la solución

| Paso | Operación | Símbolo / método |
|------|-----------|------------------|
| 2 | Intersección | `&` (equivale a `.intersection()`) |
| 3 | Unión | `\|` (equivale a `.union()`) |
| 4 | Exclusivos de A | `-` (equivale a `.difference()`) |
| 4b | Diferencia simétrica (extra) | `^` (equivale a `.symmetric_difference()`) |
| 5 | Agregar elemento | `.add()` |
| 5 | Eliminar sin error si no existe | `.discard()` |
| 6 | Conteo | `len()` |



## 7. Reflexión escrita (3 líneas)

> Un `set` es útil para evitar duplicados porque **almacena solo elementos únicos** y descarta automáticamente cualquier valor repetido al insertarlo. En datos del territorio, si se consolidan registros de municipios provenientes de varias fuentes (DANE, gobernación, alcaldías), un `set` elimina los nombres repetidos y permite comparar coberturas sin sesgos. Por ejemplo, al unir los municipios reportados por dos departamentos, `&` revela cuáles aparecen en ambos (como **Caucasia**) y `-` identifica los que solo pertenecen a uno, facilitando detectar errores o territorios compartidos.



## Detalle importante: `remove()` vs `discard()`

- `.remove("Sahagún")` lanza un `KeyError` si el elemento no existe.
- `.discard("Sahagún")` **no lanza error** si no existe, por eso se recomienda cuando no se está seguro de la presencia del elemento. En este problema ambos funcionarían porque "Sahagún" sí existe, pero `discard()` es la práctica más segura.

## Problema 4 — "De listas a DataFrame: primer vistazo a los datos" (Transición a NumPy y Pandas)

**Contexto:** Este problema conecta lo aprendido con las estructuras que se usarán en las próximas sesiones: `nd-array` y `DataFrame`.

**Enunciado:**

3. Crea un **`DataFrame`** de Pandas con dos columnas: `municipio` y `poblacion`.
4. Imprime el DataFrame completo y luego las **primeras 3 filas** con `.head(3)`.
5. Agrega una nueva columna llamada `poblacion_miles` que sea la población dividida por 1000.
6. Filtra e imprime los municipios con población **mayor a 40,000**.

# Solución — Problema 4: De listas a DataFrame


````python
"""
Problema 4 — Transición a NumPy y Pandas
Requiere: numpy y pandas  (pip install numpy pandas)
"""

import numpy as np
import pandas as pd
````

````python
# ============================================================
# Datos base (listas del Problema 2, tras agregar Cáceres y quitar Nechí)
# ============================================================
municipios = ["Caucasia", "El Bagre", "Zaragoza", "Tarazá", "Cáceres"]
poblaciones = [93843, 56197, 26996, 29542, 28944]
````

````python
# ============================================================
# 1. Crear un nd-array de NumPy con las poblaciones
# ============================================================
poblaciones_np = np.array(poblaciones)
print("--- nd-array de poblaciones ---")
print(poblaciones_np)
print(f"Tipo: {type(poblaciones_np)}")
print(f"Forma (shape): {poblaciones_np.shape}")
print(f"Tipo de dato (dtype): {poblaciones_np.dtype}")
````

````python
# ============================================================
# 2. Estadísticas con métodos de NumPy
# ============================================================
print("\n--- Estadísticas con NumPy ---")
print(f"Suma:    {poblaciones_np.sum():,}")
print(f"Promedio: {poblaciones_np.mean():,.2f}")
print(f"Máximo:  {poblaciones_np.max():,}")
print(f"Mínimo:  {poblaciones_np.min():,}")
````

````python
# ============================================================
# 3. Crear el DataFrame con dos columnas
# ============================================================
df = pd.DataFrame({
    "municipio": municipios,
    "poblacion": poblaciones,
})
print("\n--- DataFrame completo ---")
print(df)
print(f"\nDimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
````

````python
# ============================================================
# 4. Imprimir primeras 3 filas con .head(3)
# ============================================================
print("\n--- Primeras 3 filas (.head(3)) ---")
print(df.head(3))
````

````python
# ============================================================
# 5. Agregar columna poblacion_miles
# ============================================================
df["poblacion_miles"] = df["poblacion"] / 1000
print("\n--- DataFrame con columna 'poblacion_miles' ---")
print(df)
````

````python
# ============================================================
# 6. Filtrar municipios con población mayor a 40,000
# ============================================================
filtro = df[df["poblacion"] > 40000]
print("\n--- Municipios con población > 40,000 ---")
print(filtro)
````

````python
# (Extra útil) ordenar el resultado filtrado de mayor a menor
print("\n--- Filtrado y ordenado de mayor a menor ---")
print(filtro.sort_values("poblacion", ascending=True))
````

````python
# (Extra útil) mostrar formato de tabla con separador de miles
print("\n--- Filtrado con formato de miles ---")
print(filtro.to_string(index=False, formatters={"poblacion": "{:,.0f}".format,
                                                "poblacion_miles": "{:,.1f}".format}))
````

## Puntos clave de la solución

| Paso | Concepto aplicado |
|------|-------------------|
| 1 | `np.array(lista)` → convierte la lista en un `nd-array` |
| 2 | Métodos vectorizados: `.sum()`, `.mean()`, `.max()`, `.min()` |
| 3 | `pd.DataFrame({...})` con diccionario de columnas |
| 4 | `.head(3)` muestra las primeras filas (útil en datasets grandes) |
| 5 | `df["nueva"] = ...` crea columna nueva (vectorizada, sin ciclos) |
| 6 | `df[df["poblacion"] > 40000]` → filtrado booleano |
| Extra | `.sort_values()` y `.to_string(formatters=...)` para presentación |

## Equivalencias útiles lista ↔ NumPy ↔ Pandas

| Operación | Lista nativa | NumPy | Pandas |
|-----------|-------------|-------|--------|
| Suma | `sum(lista)` | `arr.sum()` | `df["col"].sum()` |
| Promedio | `sum(lista)/len(lista)` | `arr.mean()` | `df["col"].mean()` |
| Máximo | `max(lista)` | `arr.max()` | `df["col"].max()` |
| Filtrar | comprensión + `if` | máscara booleana | `df[df["col"] > x]` |
| Agregar columna | `append()` en lista paralela | `np.append()` (no ideal) | `df["nueva"] = ...` |

## 7. Reflexión escrita (3 líneas)

> Un `DataFrame` ofrece **operaciones vectorizadas** que evitan escribir ciclos explícitos (sumar, filtrar o crear columnas se hace en una línea y con mayor velocidad). Además, mantiene **columnas heterogéneas alineadas por fila** (texto y números juntos sin perder la correspondencia, a diferencia de listas paralelas) y permite **etiquetar filas y columnas**, lo que facilita seleccionar, ordenar y resumir datos. Por ejemplo, con un `DataFrame` de municipios se puede filtrar por población, agregar la densidad calculada desde área y agrupar por departamento sin reescribir la lógica de índices cada vez.
