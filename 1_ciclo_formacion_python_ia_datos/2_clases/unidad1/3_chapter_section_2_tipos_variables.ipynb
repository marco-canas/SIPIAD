## ¿Variables en Python?  


[Video de apoyo sobre creación de variables en Python](https://www.youtube.com/watch?v=U_kyVKFUlWQ)


Una variable es el nombre de una ubicación de memoria reservada que contiene un valor. 

Por ejemplo: supongamos `a = 10`. 

Aquí, "`a`" es el nombre de la variable, el signo igual (`=`) es un operador de asignación y `10` es el valor o literal. 

Por lo tanto, al usar un operador de asignación (`=`) en Python, podemos reservar memoria para un valor sin declararlo explícitamente. 

````python
edad = 51
estatura = 1.65
nombre = "Marco Cañas"
es_profesor = True 
complejo = 1 + 2j
````

````python
print(type(edad))  # print( es para pedirle que muestre el resultado )
print(type(estatura))
print(type(nombre))
print(type(es_profesor))
print(type(complejo))
````

````python
5 = x
````
