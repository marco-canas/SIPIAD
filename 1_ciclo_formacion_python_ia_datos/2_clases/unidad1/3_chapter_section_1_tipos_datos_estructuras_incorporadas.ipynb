
## 3. ¿Qué tipo usa pandas cuando una columna numérica tiene un valor faltante?



**Respuesta:** cuando una columna numérica de tipo `int` contiene al menos un valor faltante, pandas la convierte automáticamente a **`float64`**.



**¿Por qué pasa esto?**
- El valor faltante en pandas se representa como `NaN` (`Not a Number`), que internamente es un `float` especial de NumPy.
- El tipo `int64` de NumPy **no tiene forma de representar `NaN`** (no existe un "entero vacío").
- Como pandas necesita un tipo que pueda contener tanto números como el marcador de ausencia, sube todo el tipo de la columna a `float64`, que sí admite `NaN`.



**Ejemplo ilustrativo:**



````python
import pandas as pd

datos = {"humedad": [68, 72, None, 55]}
df = pd.DataFrame(datos)

print(df["humedad"].dtype)   # float64, aunque los valores "parecen" enteros
print(df)
````


**Dato extra:** desde pandas 1.0 existe también el tipo `Int64` (con "I" mayúscula, *nullable integer*), que sí permite mantener valores enteros junto con datos faltantes usando `pd.NA` en lugar de `NaN`. Pero por defecto, sin especificarlo explícitamente, pandas siempre usa `float64` para números faltantes.

# Solución: 

````python
# ============================================
# Tarea 3.1 — Inventario de una tienda
# ============================================

# --- Datos fijos de 3 productos (nombre, precio, stock, disponibilidad) ---
producto_1 = {"nombre": "Laptop",   "precio": 3_500_000.0, "stock": 5,  "disponible": True}
producto_2 = {"nombre": "Mouse",    "precio": 45000,       "stock": 30, "disponible": True}
producto_3 = {"nombre": "Teclado",  "precio": 89900.5,     "stock": 0,  "disponible": False}

productos = [producto_1, producto_2, producto_3]

# --- Procesamiento de cada producto ---
valor_inventario_total = 0.0

for p in productos:
    print(f"\nProducto: {p['nombre']}")

    # 1) Determinar el tipo de cada dato con type()
    print(f"  Tipo de nombre       -> {type(p['nombre'])}")
    print(f"  Tipo de precio       -> {type(p['precio'])}")
    print(f"  Tipo de stock        -> {type(p['stock'])}")
    print(f"  Tipo de disponible   -> {type(p['disponible'])}")

    # 3) Validar con isinstance() que el precio sea float o int
    #    (se excluye bool explícitamente, ya que bool es subtipo de int)
    precio_valido = isinstance(p["precio"], (int, float)) and not isinstance(p["precio"], bool)

    if precio_valido:
        # 2) Calcular el valor total del inventario para ese producto
        valor_producto = p["precio"] * p["stock"]
        valor_inventario_total += valor_producto
        print(f"  Precio válido -> Valor en inventario: ${valor_producto:,.2f}")
    else:
        print(f"  ❌ Precio inválido (tipo {type(p['precio'])}), no se incluye en el cálculo")

# --- Resultado final ---
print(f"\n{'='*40}")
print(f"Valor total del inventario: ${valor_inventario_total:,.2f}")
````


### Explicación de las decisiones clave



- **`type()` vs `isinstance()`**: `type()` se usó solo para *mostrar* el tipo exacto de cada dato (fines informativos), mientras que `isinstance()` se usó para *validar* el precio antes de operar, porque además acepta tuplas de tipos (`(int, float)`) y respeta la herencia — algo que `type() == float` no permite hacer de forma tan flexible.
- **Exclusión de `bool`**: en Python, `bool` es subtipo de `int` (`isinstance(True, int)` da `True`). Por eso se agregó `and not isinstance(p["precio"], bool)`, para que un valor como `precio = True` nunca se cuele como si fuera un precio válido.
- **Producto con `stock = 0`**: el Teclado tiene stock cero, por lo que su valor en inventario es `$0.00` — el cálculo sigue siendo correcto y válido, solo que no aporta al total.


> **«Por convención (PEP8), se usa `snake_case`: `tasa_interes`, `df_ventas`.»**

### 1. ¿Qué es una convención?

Una **convención** es una regla o recomendación que los programadores siguen para que el código sea **uniforme, legible y fácil de entender**.

No significa que Python obligue a escribir los nombres de esa manera. Es una **buena práctica**.



### 2. ¿Qué es PEP 8?

**PEP 8** es la guía de estilo oficial para escribir código Python de manera consistente.

Entre otras cosas, recomienda cómo nombrar:

* variables;
* funciones;
* clases;
* constantes;
* módulos.



### 3. ¿Qué significa `snake_case`?

`snake_case` es una forma de escribir nombres compuestos:

* todo en **minúsculas**;
* las palabras se separan mediante **guion bajo `_`**.

Por ejemplo:

```python
tasa_interes
df_ventas
precio_promedio
numero_estudiantes
```

Visualmente:

```text
tasa + _ + interes
 ↓        ↓
palabra  separador
```

Se llama **snake_case** porque el guion bajo hace que el nombre parezca una pequeña "serpiente" que se desplaza entre las palabras.

### 4. ¿Qué significa `df_ventas`?

Aquí hay además una convención muy utilizada en **ciencia de datos con Python**:

```python
df_ventas
```

probablemente significa:

* `df` → **DataFrame**
* `ventas` → contenido del DataFrame.

Por ejemplo:



````python
import pandas as pd

df_ventas = pd.DataFrame({
    "producto": ["A", "B", "C"],
    "ventas": [100, 150, 120]
})
````

La variable `df_ventas` permite al lector identificar rápidamente que estamos trabajando con un **DataFrame relacionado con las ventas**.

### 5. ¿Por qué no escribir `tasaInteres`?

También podrías encontrar:

```python
tasaInteres
```

Esta forma se denomina **camelCase**. Es común en otros lenguajes, pero **PEP 8 recomienda `snake_case` para variables y funciones de Python**.



Por tanto, en Python es preferible:

```python
tasa_interes
```

en lugar de:

```python
tasaInteres
```

Y también:

```python
precio_promedio
```

en lugar de:

```python
PrecioPromedio
```



### En resumen

| Forma        | Ejemplo        | Uso en Python                             |
| ------------ | -------------- | ----------------------------------------- |
| `snake_case` | `tasa_interes` | ✅ Recomendado para variables y funciones  |
| `camelCase`  | `tasaInteres`  | ⚠️ No es la convención habitual de Python |
| `PascalCase` | `TasaInteres`  | ✅ Habitual para clases                    |
| MAYÚSCULAS   | `TASA_INTERES` | ✅ Habitual para constantes                |



**La idea fundamental es:** cuando programes en Python, procura que los nombres de variables y funciones sean **descriptivos, en minúsculas y con palabras separadas por `_`**. Esto hace que tu código sea mucho más fácil de leer, especialmente en proyectos de análisis de datos y Machine Learning.

## 3. ¿Por qué una variable es "una referencia" y no "una caja"?



En Python, una variable no almacena el valor directamente en un espacio de memoria propio; en realidad es una **etiqueta que apunta** a un objeto que ya existe en memoria. 

Por eso, al hacer `b = a`, ambas variables terminan apuntando **al mismo objeto** (mismo `id()`), y no a dos copias independientes como sugeriría la metáfora de la "caja".

Crea un script que use asignación múltiple para inicializar los hiperparámetros de un modelo (`learning_rate`, `epochs`, `batch_size`) y luego los reasigne dentro de un ciclo `for` que simule 3 configuraciones distintas, imprimiendo cada combinación.

---


````python
# ============================================
# Tarea 3.2 — Hiperparámetros de un modelo
# ============================================

# --- Inicialización con asignación múltiple en una sola línea ---
learning_rate, epochs, batch_size = 0.01, 10, 32

print("Configuración inicial:")
print(f"  learning_rate={learning_rate} | epochs={epochs} | batch_size={batch_size}\n")

# --- Lista de 3 configuraciones distintas a simular ---
configuraciones = [
    (0.001, 20, 16),
    (0.01,  50, 32),
    (0.1,   100, 64),
]

# --- Ciclo for que reasigna las variables en cada iteración ---
for i, (learning_rate, epochs, batch_size) in enumerate(configuraciones, start=1):
    print(f"Configuración {i}:")
    print(f"  learning_rate={learning_rate} | epochs={epochs} | batch_size={batch_size}")
    print("-" * 45)
````


### Explicación de las decisiones clave

- **Asignación múltiple inicial:** `learning_rate, epochs, batch_size = 0.01, 10, 32` crea las tres variables en una sola línea, cada una apuntando a su propio valor (`float`, `int`, `int`).
- **Reasignación dentro del ciclo:** en cada vuelta del `for`, la tupla `(learning_rate, epochs, batch_size) = configuraciones[i]` se **desempaqueta** directamente sobre las mismas tres variables, reasignándolas por completo — esto es posible gracias al tipado dinámico de Python (las variables no están "atadas" a un tipo ni a un valor fijo).
- **`enumerate(..., start=1)`:** se usó para numerar las configuraciones desde 1 en la impresión, sin necesidad de llevar un contador manual con `+= 1`.
- **Uso de tuplas para las configuraciones:** cada configuración se representó como una tupla inmutable (`(lr, epochs, batch)`), ya que son valores fijos que no deberían modificarse una vez definidos — coherente con lo visto en la sección 3.3.2 sobre tuplas.


# 3.3 Estructuras de Datos Incorporadas en Python (Mutables e Inmutables)



En Python, cada estructura de datos se clasifica según si su contenido puede modificarse **después de creado**:



| Categoría | Estructuras | ¿Se puede modificar? |

| **Inmutables** | `tuple`, `str`, `int`, `float` | No, cualquier "cambio" crea un nuevo objeto |

- **Lista:** Al modificar un elemento, el **ID no cambia** porque la lista es **mutable** (se modifica el objeto existente).

### 3. Reflexión: ¿Por qué las tuplas son más seguras para nombres de columnas de un dataset?

**Respuesta:**

Las tuplas son más seguras porque:

1. **Inmutabilidad:** Al ser inmutables, una vez definidos los nombres de las columnas, **no se pueden modificar accidentalmente**. Esto evita errores como:
   - Renombrar una columna sin querer.
   - Sobrescribir un nombre con otro valor.
   - Introducir bugs difíciles de rastrear.



2. **Integridad de los datos:** En análisis de datos, los nombres de las columnas son la "clave" para acceder a los datos. Si cambian, todo el código que depende de ellos puede romperse.

3. **Hashable:** Las tuplas son *hashable* (se pueden usar como claves de diccionarios), lo que permite usarlas en estructuras como `dict` o `set` para validaciones rápidas. Las listas no lo son.

4. **Intención clara:** Usar una tupla comunica a otros programadores: *"Estos nombres no deben cambiar"*, haciendo el código más legible y mantenible.


### Comparación detallada: `sorted()` vs `.sort()`

| Característica | `sorted()` | `.sort()` |
|----------------|------------|-----------|
| **Modifica la original** | ❌ No, crea una nueva | ✅ Sí, modifica in-place |
| **Retorna valor** | ✅ Nueva lista ordenada | ❌ None (modifica el objeto) |
| **Uso típico** | Cuando necesitas mantener la original intacta | Cuando ya no necesitas el orden original |
| **Sintaxis** | `nueva = sorted(lista)` | `lista.sort()` |
| **Memoria** | Crea una copia (más memoria) | No crea copia (menos memoria) |
| **Aplicable a** | Cualquier iterable (tuplas, sets, etc.) | Solo listas |

### Situación 1 — El formulario mal diseñado
**Contexto:** Una startup recolectó datos de 500 usuarios mediante un formulario web mal configurado: todos los campos (edad, ingresos, ¿acepta términos?) llegaron como texto (`str`), incluso los numéricos y booleanos.
**Reto:** Diseña una estrategia para detectar y convertir automáticamente cada campo al tipo correcto, manejando los casos en que la conversión falle (por ejemplo, un usuario escribió "veintiocho" en vez de "28").
**Preguntas orientadoras:**
- ¿Cómo decides qué tipo "debería" tener cada campo?
- ¿Qué haces con los valores que no se pueden convertir?
**Entregable:** Script que reciba una lista de diccionarios con valores en `str` y produzca una nueva lista con los tipos corregidos, más un reporte de cuántos registros no pudieron convertirse.



Aquí tienes una solución completa en Python, con una estrategia clara y un script funcional.

## Estrategia de diseño

**1. ¿Cómo decido qué tipo "debería" tener cada campo?**
Defino un **esquema de tipos esperados** (un diccionario `campo -> tipo`), porque el nombre del campo no es suficiente evidencia por sí solo (ej. `"edad"` podría venir como `"28"`, `"28.0"` o `"veintiocho"`). Este esquema es la fuente de verdad de negocio, no algo que se infiera mágicamente del contenido.

**2. ¿Qué hago con los valores que no se pueden convertir?**
- No descarto el registro completo: marco el **campo específico** como inválido (`None`) y guardo el motivo.
- Genero un **reporte de errores** (registro, campo, valor original, motivo).
- Cuento cuántos **registros** tienen al menos un fallo, y cuántos **campos individuales** fallaron.
- Uso funciones de conversión "tolerantes" para booleanos (acepta `"sí"`, `"si"`, `"true"`, `"1"`, `"no"`, etc.) antes de rendirme.



````python
from dataclasses import dataclass, field
from typing import Any, Callable
from datetime import datetime

# ---------------------------------------------------------
# 1. ESQUEMA: qué tipo "debería" tener cada campo
# ---------------------------------------------------------
SCHEMA = {
    "edad": int,
    "ingresos": float,
    "acepta_terminos": bool,
}

# Representaciones aceptadas de booleanos en texto libre
VERDADEROS = {"true", "si", "sí", "1", "yes", "verdadero", "s"}
FALSOS = {"false", "no", "0", "n", "falso"}


def convertir_bool(valor: str) -> bool:
    v = valor.strip().lower()
    if v in VERDADEROS:
        return True
    if v in FALSOS:
        return False
    raise ValueError(f"No se reconoce '{valor}' como booleano")


def convertir_int(valor: str) -> int:
    v = valor.strip().replace(",", "") # Aquí se esta suponiendo que la coma la están utilizando como punto que indica unidades de mil
    return int(float(v))  # tolera "28.0" -> 28


def convertir_float(valor: str) -> float:
    v = valor.strip().replace(",", ".")
    return float(v)


CONVERSORES: dict[type, Callable[[str], Any]] = {
    int: convertir_int,
    float: convertir_float,
    bool: convertir_bool,
}


@dataclass
class ReporteConversion:
    total_registros: int = 0
    registros_con_error: int = 0
    campos_fallidos: int = 0
    errores: list[dict] = field(default_factory=list)

    def resumen(self) -> str:
        return (
            f"--- Reporte de conversión ---\n"
            f"Total de registros procesados : {self.total_registros}\n"
            f"Registros con al menos 1 error : {self.registros_con_error}\n"
            f"Campos individuales fallidos   : {self.campos_fallidos}\n"
            f"Detalle de errores             : {len(self.errores)} entradas\n"
        )


def convertir_registros(
    datos: list[dict[str, str]],
    esquema: dict[str, type] = SCHEMA,
) -> tuple[list[dict[str, Any]], ReporteConversion]:
    """
    Recibe una lista de diccionarios con valores str y devuelve:
      - una nueva lista con los tipos corregidos (o None si falló)
      - un reporte de la conversión
    """
    reporte = ReporteConversion(total_registros=len(datos))
    resultado = []

    for i, registro in enumerate(datos):
        nuevo_registro = {}
        tuvo_error = False

        for campo, valor in registro.items():
            tipo_esperado = esquema.get(campo)

            # Si el campo no está en el esquema, se deja tal cual (str)
            if tipo_esperado is None:
                nuevo_registro[campo] = valor
                continue

            conversor = CONVERSORES[tipo_esperado]
            try:
                nuevo_registro[campo] = conversor(valor)
            except (ValueError, TypeError) as e:
                nuevo_registro[campo] = None
                tuvo_error = True
                reporte.campos_fallidos += 1
                reporte.errores.append({
                    "registro_index": i,
                    "campo": campo,
                    "valor_original": valor,
                    "motivo": str(e),
                })

        if tuvo_error:
            reporte.registros_con_error += 1

        resultado.append(nuevo_registro)

    return resultado, reporte


# ---------------------------------------------------------
# DEMO
# ---------------------------------------------------------
if __name__ == "__main__":
    datos_crudos = [
        {"edad": "28", "ingresos": "1500.50", "acepta_terminos": "true"},
        {"edad": "veintiocho", "ingresos": "2000", "acepta_terminos": "si"},
        {"edad": "35", "ingresos": "mil quinientos", "acepta_terminos": "no"},
        {"edad": "40", "ingresos": "3200,75", "acepta_terminos": "tal vez"},
    ]

    datos_limpios, reporte = convertir_registros(datos_crudos)

    print("Datos convertidos:")
    for r in datos_limpios:
        print(r)

    print()
    print(reporte.resumen())

    print("Detalle de errores:")
    for err in reporte.errores:
        print(err)
````

{'edad': 35, 'ingresos': None, 'acepta_terminos': False}
{'edad': 40, 'ingresos': 3200.75, 'acepta_terminos': None}

### Puntos clave de la solución
- **Separación de responsabilidades**: el esquema define *qué* tipo se espera; los conversores definen *cómo* convertir; la función principal orquesta y registra errores.
- **Tolerancia progresiva**: antes de fallar, se prueban variantes razonables (comas decimales, "sí/no", espacios).
- **Trazabilidad**: cada fallo queda documentado con índice de registro, campo, valor original y motivo — útil para depurar el formulario original o pedir corrección al usuario.
- **Extensible**: agregar un nuevo campo al `SCHEMA` (por ejemplo `"fecha_registro": datetime`) solo requiere añadir su conversor a `CONVERSORES`.
