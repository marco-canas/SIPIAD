### Módulo 1 — Tipos de datos primitivos y operadores

```python
# M1_tipos_datos.ipynb

# --- Tipos de datos primitivos ---
entero = 42
flotante = 3.1416
cadena = "SIPIAD - Universidad de Antioquia"
booleano = True
nulo = None

print(type(entero), entero)
print(type(flotante), flotante)

print("División:", a / b)
print("División entera:", a // b)

print("Potencia:", a ** b)

# --- Operadores lógicos ---
x, y = True, False
print("AND:", x and y)
print("OR:", x or y)
print("NOT:", not x)

# --- Conversión de tipos (casting) ---
print(int("2026"))
print(float("3.14"))
print(str(17))
print(bool(0), bool(1))
```

### Módulo 2 — Listas y tuplas

```python
# M2_listas_tuplas.ipynb

# --- LISTAS (mutables) ---
municipios_bajo_cauca = ["Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá", "Cáceres"]
print("Lista original:", municipios_bajo_cauca)

# Indexación y slicing
print("Primer municipio:", municipios_bajo_cauca[0])
print("Último municipio:", municipios_bajo_cauca[-1])
print("Primeros tres:", municipios_bajo_cauca[:3])
print("Desde el tercero:", municipios_bajo_cauca[2:])

# Modificación
municipios_bajo_cauca.append("Cáceres")  # (ya estaba, pero se permite duplicado en listas)
municipios_bajo_cauca[0] = "Caucasia (Antioquia)"
print("Lista modificada:", municipios_bajo_cauca)

# Métodos útiles
print("Longitud:", len(municipios_bajo_cauca))
print("¿Contiene 'Nechí'?:", "Nechí" in municipios_bajo_cauca)
print("Índice de 'Zaragoza':", municipios_bajo_cauca.index("Zaragoza"))
municipios_bajo_cauca.remove("Cáceres")
print("Sin Cáceres:", municipios_bajo_cauca)

# --- TUPLAS (inmutables) ---
coordenadas_caucasia = (-7.9833, -75.2000)
print("Coordenadas:", coordenadas_caucasia)
print("Latitud:", coordenadas_caucasia[0])

# Intento de modificación (genera error, se muestra comentado)
# coordenadas_caucasia[0] = -8.0  # TypeError

# Desempaquetado
lat, lon = coordenadas_caucasia
print(f"Latitud = {lat}, Longitud = {lon}")

# --- Listas vs. Tuplas ---
# Las listas se usan para datos que cambian; las tuplas para datos fijos.
```

### Módulo 3 — Sets y diccionarios


````python
# M3_sets_diccionarios.ipynb

# --- SETS (conjuntos: únicos, no ordenados) ---
municipios_antioquia = {"Caucasia", "El Bagre", "Medellín", "Nechí", "Zaragoza"}
municipios_cordoba = {"Montería", "Cereté", "Caucasia", "Sahagún"}

print("Antioquia:", municipios_antioquia)
print("Córdoba:", municipios_cordoba)

# Operaciones de conjuntos
print("Intersección:", municipios_antioquia & municipios_cordoba)
print("Unión:", municipios_antioquia | municipios_cordoba)
print("Diferencia (A - B):", municipios_antioquia - municipios_cordoba)
print("Diferencia simétrica:", municipios_antioquia ^ municipios_cordoba)

# Agregar y eliminar
municipios_antioquia.add("Tarazá")
municipios_antioquia.discard("Medellín")
print("Actualizado:", municipios_antioquia)

# --- DICCIONARIOS (clave-valor) ---
poblacion_municipios = {
    "Caucasia": 122000,
    "El Bagre": 60000,
    "Zaragoza": 32000,
    "Nechí": 30000,
    "Tarazá": 50000
}

print("Población de Caucasia:", poblacion_municipios["Caucasia"])

# Métodos útiles
print("Claves:", list(poblacion_municipios.keys()))
print("Valores:", list(poblacion_municipios.values()))
print("Ítems:", list(poblacion_municipios.items()))

# Agregar / modificar
poblacion_municipios["Cáceres"] = 40000
poblacion_municipios["Zaragoza"] = 33000  # actualización
print("Diccionario actualizado:", poblacion_municipios)

# Recorrido
for municipio, poblacion in poblacion_municipios.items():
    print(f"{municipio}: {poblacion:,} habitantes")

# Búsqueda segura
print("Población de 'Medellín':", poblacion_municipios.get("Medellín", "No registrado"))

````


### Módulo 4 — Transición a `nd-array` y `DataFrame`

```python
# M4_numpy_pandas.ipynb
import numpy as np
import pandas as pd

# --- ND-ARRAY (NumPy) ---
# Creación de un array unidimensional
poblaciones = np.array([122000, 60000, 32000, 30000, 50000])
print("Array 1D:", poblaciones)
print("Tipo:", type(poblaciones))
print("Dimensiones:", poblaciones.ndim)
print("Forma:", poblaciones.shape)

# Operaciones vectorizadas (ventaja clave de NumPy)
print("Población total:", poblaciones.sum())
print("Promedio:", poblaciones.mean())
print("Máximo:", poblaciones.max())
print("Mínimo:", poblaciones.min())
print("Poblaciones en miles:", poblaciones / 1000)

# Array 2D
matriz_datos = np.array([
    [122000, 2023],
    [60000, 2023],
    [32000, 2023],
    [30000, 2023],
    [50000, 2023]
])
print("Array 2D:\n", matriz_datos)
print("Forma:", matriz_datos.shape)

# --- DATAFRAME (Pandas) ---
datos = {
    "municipio": ["Caucasia", "El Bagre", "Zaragoza", "Nechí", "Tarazá"],
    "poblacion": [122000, 60000, 32000, 30000, 50000],
    "departamento": ["Antioquia", "Antioquia", "Antioquia", "Antioquia", "Antioquia"]
}
df = pd.DataFrame(datos)
print("\nDataFrame:\n", df)

# Exploración básica
print("\nPrimeras filas:\n", df.head())
print("\nInformación:\n", df.info())
print("\nEstadísticas descriptivas:\n", df.describe())

# Selección de columnas
print("\nColumna 'municipio':\n", df["municipio"])

# Filtrado
print("\nMunicipios con más de 50,000 habitantes:\n", df[df["poblacion"] > 50000])

# Agregar columna
df["poblacion_miles"] = df["poblacion"] / 1000
print("\nDataFrame con columna calculada:\n", df)

# --- Comparación: lista vs. nd-array vs. DataFrame ---
lista_nativa = [122000, 60000, 32000, 30000, 50000]
array_np = np.array(lista_nativa)
print("\nLista nativa:", lista_nativa)
print("Nd-array:", array_np)
print("¿Mismo contenido?:", list(array_np) == lista_nativa)
```
