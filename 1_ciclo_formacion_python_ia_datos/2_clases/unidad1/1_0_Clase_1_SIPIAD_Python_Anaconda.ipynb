## 2. Una breve historia de Python

Python fue creado por **Guido van Rossum** a finales de los años ochenta y comenzó a publicarse en 1991. El diseño buscaba favorecer un código legible y una experiencia de programación práctica. El nombre se inspiró en el grupo humorístico *Monty Python*, no en la serpiente.

Hitos generales:

- **1991:** primeras publicaciones públicas de Python.
- **2000:** Python 2.0 incorpora, entre otras mejoras, recolección de basura y soporte para listas por comprensión.
- **2008:** se publica Python 3.0, con cambios importantes que no eran totalmente compatibles con Python 2.
- **2020:** finaliza el soporte oficial de Python 2.
- **Actualidad:** Python se usa en desarrollo de software, automatización, educación, análisis de datos, computación científica e IA. El ecosistema evoluciona continuamente.

La historia no es una lista para memorizar: muestra cómo un lenguaje cambia junto con las necesidades de quienes lo utilizan y de las comunidades que mantienen sus bibliotecas.

**Idea clave:** Python es un proyecto de software libre y de código abierto desarrollado por una comunidad internacional. La Python Software Foundation apoya y administra aspectos centrales del proyecto.

## 3. Diferentes versiones de Python

### 3.1 ¿Cómo se interpreta un número de versión?

Una versión como `3.14.8` se puede leer, de manera general, así:

- `3`: versión mayor o familia principal.
- `14`: versión menor.
- `8`: versión de mantenimiento o corrección.

El esquema exacto y el calendario dependen de las políticas de publicación del proyecto. Las versiones preliminares (alpha, beta o release candidate) se usan para probar cambios y no suelen ser la primera opción para un entorno de investigación estable.

### 3.2 Versiones y compatibilidad

No siempre conviene instalar “la versión más nueva” sin revisar las necesidades del proyecto. Antes de elegir, verifica:
- que la versión tenga soporte;
- que las bibliotecas necesarias sean compatibles;
- que el sistema operativo y la arquitectura estén soportados;
- que el equipo documente la versión y las dependencias.

A fecha de preparación de esta clase (**20 de octubre de 2026**), el sitio oficial de Python presenta Python 3.15.0 como versión estable reciente y Python 3.14 como otra serie activa. La disponibilidad puede cambiar: consulta la página oficial antes de instalar.

**Fuentes oficiales:**
- [Descargas y versiones de Python](https://www.python.org/downloads/)
- [Ciclo de vida de versiones](https://devguide.python.org/versions/)
- [Novedades de Python](https://docs.python.org/3/whatsnew/)

````python
import sys
import platform

print("Versión de Python:", sys.version)
print("Versión resumida:", platform.python_version())
print("Sistema operativo:", platform.system())
print("Arquitectura:", platform.machine())
````

**Interpretación:** ejecuta la celda y registra la versión, el sistema operativo y la arquitectura. Estos datos son parte de la descripción del entorno computacional de un experimento.

## 4. Características de Python

### Legibilidad
* La sintaxis favorece código que se puede leer y discutir en equipo.   
* La indentación delimita bloques de código.

### Lenguaje de alto nivel
Permite expresar ideas sin administrar directamente muchos detalles de bajo nivel del hardware.

### Multiparadigma
Permite programar con funciones, objetos y otros estilos, según las necesidades del problema.

### Tipado dinámico
Los nombres pueden referirse a valores de distintos tipos durante la ejecución. Esto agiliza la exploración, pero exige buenas pruebas y claridad.

### Ecosistema de bibliotecas
- **NumPy:** arreglos y operaciones numéricas.
- **pandas:** tablas y manipulación de datos.
- **Matplotlib / Seaborn / Plotly:** visualización.
- **SciPy / SymPy:** computación científica y matemática simbólica.
- **statsmodels / scikit-learn:** estadística y aprendizaje automático.
- **PyTorch / TensorFlow:** aprendizaje profundo.
- **Jupyter:** documentos interactivos que combinan código, texto, ecuaciones y resultados.

### Portabilidad y comunidad
Python está disponible en los principales sistemas operativos y cuenta con documentación, foros y proyectos abiertos. Aun así, la portabilidad depende de las bibliotecas, versiones y configuraciones utilizadas.

### Integración con investigación reproducible
Un cuaderno puede reunir pregunta, método, código, resultados y discusión. Para reproducibilidad se deben documentar datos, versiones, dependencias, semillas aleatorias y pasos de procesamiento. Un cuaderno por sí solo no garantiza que un resultado sea reproducible.

````python
# Ejemplo: legibilidad, variables, operaciones y una función
temperaturas_c = [28.5, 29.0, 30.2, 27.8, 31.1]

def promedio(valores):
    return sum(valores) / len(valores)

media = promedio(temperaturas_c)
print(f"Cantidad de observaciones: {len(temperaturas_c)}")
print(f"Temperatura media: {media:.2f} °C")
````

**Lectura crítica del ejemplo:** estos valores son ficticios y solo ilustran operaciones con Python. No representan mediciones reales de Caucasia. En investigación debemos distinguir con claridad los datos simulados de los observados.

### Mini-experimento: datos sintéticos para aprender

Imaginemos una serie corta de lluvia diaria. Los números de la siguiente celda son **simulados**, no datos del IDEAM.

````python
import pandas as pd
import matplotlib.pyplot as plt

datos = pd.DataFrame({
    "dia": range(1, 11),
    "lluvia_mm": [0, 2, 0, 12, 8, 0, 4, 18, 7, 1]
})

display(datos)
````

````python
ax = datos.plot(x="dia", y="lluvia_mm", marker="o", legend=False)
ax.set_title("Lluvia diaria simulada")
ax.set_xlabel("Día")
ax.set_ylabel("Lluvia (mm)")
plt.tight_layout()
plt.show()
````

**Preguntas de investigación iniciales**
1. ¿Qué día se registró el mayor valor simulado?
2. ¿Cuántos días tuvieron lluvia igual a cero?
3. ¿Qué información adicional necesitaríamos para analizar una temporada de lluvias real?
4. ¿Qué diferencia habría entre describir esta serie y predecir el día siguiente?

````python
dia_maximo = datos.loc[datos["lluvia_mm"].idxmax(), "dia"]
maximo = datos["lluvia_mm"].max()
dias_secos = (datos["lluvia_mm"] == 0).sum()

print(f"Día con mayor lluvia simulada: {dia_maximo}")
print(f"Máximo simulado: {maximo} mm")
print(f"Días con lluvia igual a cero: {dias_secos}")
````

## 6. ¿Qué es Anaconda Distribution?

**Anaconda Distribution** es una distribución para trabajar con Python que incluye herramientas como Python, Jupyter y Anaconda Navigator, además de un conjunto amplio de paquetes orientados a ciencia de datos. Incluye **conda**, una herramienta para administrar entornos y paquetes.

Conceptos que no debemos confundir:

- **Python:** lenguaje e intérprete.
- **Paquete o biblioteca:** código reutilizable, por ejemplo `pandas`.
- **Entorno:** conjunto aislado de Python y dependencias para un proyecto.
- **conda:** herramienta para administrar entornos y paquetes.
- **Anaconda Distribution:** distribución con herramientas y paquetes preinstalados.
- **Anaconda Navigator:** interfaz gráfica para abrir aplicaciones y gestionar entornos.
- **Jupyter Notebook / JupyterLab:** herramientas para trabajar con documentos interactivos.

### ¿Por qué usarla en SIPIAD?
Permite comenzar con un conjunto de herramientas científicas sin instalar cada biblioteca por separado. Para proyectos posteriores, los entornos independientes ayudan a evitar conflictos entre dependencias.

**Antes de descargar:** Anaconda requiere espacio de almacenamiento considerable; revisa los requisitos actuales. La distribución y sus repositorios tienen términos de servicio que pueden variar según el uso y el tipo de organización. Consulta las condiciones vigentes de Anaconda, especialmente si el uso se realiza en una institución.

Cuando el proyecto crezca, conviene crear un entorno independiente. Los siguientes comandos se ejecutan en **Anaconda Prompt**, no en una celda Python normal.

Para salir del entorno:

```text
conda deactivate
```

Para ver los entornos existentes:

```text
conda env list
```

### Reproducibilidad mínima
Para compartir el entorno del proyecto, puedes exportarlo desde Anaconda Prompt:

```text
conda env export --from-history > environment.yml
```

El archivo `environment.yml` registra dependencias solicitadas explícitamente, aunque puede no capturar todos los detalles necesarios para reproducir idénticamente el sistema en otra máquina. Documenta también sistema operativo, versiones relevantes, fuentes de datos y pasos de preparación.

## 9. Primer registro del entorno computacional

Ejecuta esta celda después de instalar Anaconda y abrir el cuaderno en el entorno elegido.
