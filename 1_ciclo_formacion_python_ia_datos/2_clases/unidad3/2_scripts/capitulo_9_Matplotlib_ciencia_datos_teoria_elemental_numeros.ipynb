## 4. Condicionales `if`, `elif`, `else`: clasificar observaciones

Primero clasificamos una nota individual. Luego aplicamos la función a una columna de Pandas.

````python
nota = 3.7

if nota >= 4.5:
    nivel = "Alto"
elif nota >= 3.0:
    nivel = "Aprobado"
else:
    nivel = "Debe mejorar"

print(nota, "->", nivel)
````

````python
def clasificar_nota(nota):
    if nota >= 4.5:
        return "Alto"
    elif nota >= 3.0:
        return "Aprobado"
    else:
        return "Debe mejorar"

datos_estudiantes["nivel"] = datos_estudiantes["nota"].apply(clasificar_nota)
conteo_niveles = datos_estudiantes["nivel"].value_counts()
display(datos_estudiantes)
display(conteo_niveles)

plt.figure(figsize=(7, 4))
conteo_niveles.plot(kind="bar")
plt.xlabel("Nivel")
plt.ylabel("Cantidad de estudiantes")
plt.title("Cantidad de estudiantes por nivel")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
````

## 5. Teoría elemental de números: divisibilidad

El operador módulo `%` permite reconocer múltiplos. NumPy crea los enteros y produce máscaras booleanas; Pandas organiza los resultados en columnas.

````python
numeros = pd.DataFrame({"n": np.arange(1, 51)})
numeros["par"] = numeros["n"] % 2 == 0
numeros["multiplo_3"] = numeros["n"] % 3 == 0
numeros["multiplo_5"] = numeros["n"] % 5 == 0
numeros["multiplo_3_y_5"] = numeros["multiplo_3"] & numeros["multiplo_5"]

display(numeros.head(15))
print("Múltiplos de 3:", numeros.loc[numeros["multiplo_3"], "n"].tolist())
print("Múltiplos de 3 y 5:", numeros.loc[numeros["multiplo_3_y_5"], "n"].tolist())
````

````python
plt.figure(figsize=(9, 4))
plt.scatter(numeros["n"], numeros["n"] % 3, label="Residuo al dividir entre 3")
plt.xlabel("n")
plt.ylabel("n % 3")
plt.title("Patrón de residuos módulo 3")
plt.yticks([0, 1, 2])
plt.legend()
plt.grid(True)
plt.show()
````

## 6. Ciclo `for`: contar divisores

Un divisor positivo de `n` es un entero positivo que divide a `n` sin dejar residuo. Usamos un ciclo `for` para implementar la definición.

````python
def divisores_de(n):
    divisores = []
    for d in range(1, n + 1):
        if n % d == 0:
            divisores.append(d)
    return divisores

for n in [12, 18, 25]:
    print(f"Divisores de {n}: {divisores_de(n)}")
````

### 6.1 Cantidad de divisores para cada entero

Aplicamos la función a una columna de Pandas y representamos la cantidad de divisores.

````python
tabla_divisores = pd.DataFrame({"n": np.arange(1, 101)})
tabla_divisores["divisores"] = tabla_divisores["n"].apply(divisores_de)
tabla_divisores["cantidad_divisores"] = tabla_divisores["divisores"].apply(len)

display(tabla_divisores.head(10))

plt.figure(figsize=(10, 4))
plt.scatter(tabla_divisores["n"], tabla_divisores["cantidad_divisores"], s=25)
plt.xlabel("Entero n")
plt.ylabel("Cantidad de divisores positivos")
plt.title("Cantidad de divisores de los enteros del 1 al 100")
plt.grid(True)
plt.show()
````

## 7. Ciclo `while`: detectar números primos

Un número primo es un entero mayor que 1 cuyos únicos divisores positivos son 1 y él mismo. Si `n` es compuesto, tiene un divisor no mayor que su raíz cuadrada.

````python
def es_primo(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

for n in [1, 2, 9, 17, 21, 29]:
    print(n, "es primo:", es_primo(n))
````

### 7.1 Visualizar la distribución de primos

Creamos una columna booleana y usamos una gráfica de dispersión para marcar los primos.

````python
primos_df = pd.DataFrame({"n": np.arange(1, 501)})
primos_df["es_primo"] = primos_df["n"].apply(es_primo)
primos_df["indicador_primo"] = primos_df["es_primo"].astype(int)

display(primos_df.head(15))
print("Cantidad de primos entre 1 y 500:", primos_df["es_primo"].sum())

plt.figure(figsize=(10, 3))
plt.scatter(primos_df["n"], primos_df["indicador_primo"], s=18)
plt.xlabel("n")
plt.ylabel("¿Es primo? (1 = sí, 0 = no)")
plt.title("Distribución de los números primos hasta 500")
plt.yticks([0, 1], ["No primo", "Primo"])
plt.grid(True)
plt.show()
````

## 8. Funciones `def` y `lambda`

`def` es adecuado para funciones reutilizables y algoritmos con varias instrucciones. `lambda` resulta útil para una transformación breve.

````python
def ultimo_digito(n):
    return n % 10

es_par = lambda n: n % 2 == 0
termina_en_5 = lambda n: n % 10 == 5

ejemplos = pd.DataFrame({"n": [12, 25, 37, 40, 58, 63, 75, 82]})
ejemplos["ultimo_digito"] = ejemplos["n"].apply(ultimo_digito)
ejemplos["es_par"] = ejemplos["n"].apply(es_par)
ejemplos["termina_en_5"] = ejemplos["n"].apply(termina_en_5)
display(ejemplos)
````

### 8.1 Visualizar frecuencias del último dígito

````python
todos = pd.DataFrame({"n": np.arange(1, 501)})
todos["ultimo_digito"] = todos["n"].apply(lambda x: x % 10)
frecuencias = todos["ultimo_digito"].value_counts().sort_index()

plt.figure(figsize=(8, 4))
plt.bar(frecuencias.index, frecuencias.values)
plt.xlabel("Último dígito")
plt.ylabel("Frecuencia")
plt.title("Frecuencia de los últimos dígitos entre 1 y 500")
plt.xticks(range(10))
plt.grid(axis="y")
plt.show()
````

## 9. Ciencia de datos: simular mediciones y detectar valores atípicos

Usaremos NumPy para generar datos, Pandas para organizarlos y Matplotlib para visualizar su distribución. La semilla hace reproducible la simulación.

````python
rng = np.random.default_rng(2026)
mediciones = np.round(rng.normal(loc=20, scale=2.5, size=200), 2)
mediciones[10] = 35
mediciones[80] = 8

sensor = pd.DataFrame({
    "observacion": np.arange(1, len(mediciones) + 1),
    "medicion": mediciones
})
sensor["atipica"] = (sensor["medicion"] > 25) | (sensor["medicion"] < 12)

display(sensor.head())
display(sensor["medicion"].describe())
print("Observaciones atípicas:", int(sensor["atipica"].sum()))
````

````python
plt.figure(figsize=(8, 4))
plt.hist(sensor["medicion"], bins=20, edgecolor="black")
plt.xlabel("Medición")
plt.ylabel("Frecuencia")
plt.title("Distribución de las mediciones simuladas")
plt.grid(axis="y")
plt.show()
````

````python
plt.figure(figsize=(10, 4))
plt.plot(sensor["observacion"], sensor["medicion"], marker=".", linestyle="none", label="Mediciones")
atipicas = sensor[sensor["atipica"]]
plt.scatter(atipicas["observacion"], atipicas["medicion"], label="Atípicas")
plt.axhline(25, linestyle="--", label="Umbral superior")
plt.axhline(12, linestyle="--", label="Umbral inferior")
plt.xlabel("Número de observación")
plt.ylabel("Medición")
plt.title("Detección visual de valores fuera de los umbrales")
plt.legend()
plt.grid(True)
plt.show()
````

## 10. Comparar categorías con `if`, `elif` y `else`

Definimos una función para clasificar mediciones. Luego calculamos frecuencias con Pandas y las visualizamos con Matplotlib.

````python
def clasificar_medicion(x):
    if x < 18:
        return "baja"
    elif x <= 23:
        return "habitual"
    else:
        return "alta"

sensor["categoria"] = sensor["medicion"].apply(clasificar_medicion)
conteos = sensor["categoria"].value_counts().reindex(["baja", "habitual", "alta"], fill_value=0)
display(conteos)

plt.figure(figsize=(7, 4))
plt.bar(conteos.index, conteos.values)
plt.xlabel("Categoría")
plt.ylabel("Número de mediciones")
plt.title("Clasificación de las mediciones")
plt.grid(axis="y")
plt.show()
````

## 11. Explorar una conjetura con datos: brechas entre primos

Una brecha entre primos consecutivos es la diferencia entre un primo y el primo anterior. Las visualizaciones ayudan a explorar patrones, pero un número finito de ejemplos no prueba una conjetura para todos los enteros.

````python
lista_primos = primos_df.loc[primos_df["es_primo"], "n"].to_numpy()
brechas = np.diff(lista_primos)

brechas_df = pd.DataFrame({
    "primo_anterior": lista_primos[:-1],
    "primo_siguiente": lista_primos[1:],
    "brecha": brechas
})
display(brechas_df.head(15))

plt.figure(figsize=(10, 4))
plt.scatter(brechas_df["primo_anterior"], brechas_df["brecha"], s=22)
plt.xlabel("Primo anterior")
plt.ylabel("Brecha al siguiente primo")
plt.title("Brechas entre primos consecutivos hasta 500")
plt.grid(True)
plt.show()
````

````python
# Ejercicio 6: suma de dígitos
def suma_digitos(n):
    n = abs(int(n))
    suma = 0
    while n > 0:
        suma += n % 10
        n //= 10
    return suma

digitos = pd.DataFrame({"n": np.arange(1, 101)})
digitos["suma_digitos"] = digitos["n"].apply(suma_digitos)

plt.figure(figsize=(9, 4))
plt.scatter(digitos["n"], digitos["suma_digitos"], s=25)
plt.xlabel("n")
plt.ylabel("Suma de sus dígitos")
plt.title("Suma de dígitos de los enteros del 1 al 100")
plt.grid(True)
plt.show()
````
