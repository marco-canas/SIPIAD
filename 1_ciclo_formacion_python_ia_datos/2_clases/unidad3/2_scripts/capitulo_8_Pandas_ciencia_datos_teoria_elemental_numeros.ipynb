## 2. Crear una `Series` y un `DataFrame`

Una `Series` es una secuencia etiquetada de datos. Un `DataFrame` es una tabla con filas y columnas.

````python
notas = pd.Series([3.2, 4.1, 2.8, 4.7, 3.9], name="nota")
print(notas)
print("Promedio:", notas.mean())

estudiantes = pd.DataFrame({
    "estudiante": ["Ana", "Luis", "Marta", "José", "Sofía"],
    "edad": [18, 20, 19, 21, 18],
    "nota": notas.to_numpy()
})
display(estudiantes)
print("Dimensiones:", estudiantes.shape)
print("Columnas:", estudiantes.columns.tolist())
print("Tipos:", estudiantes.dtypes)
````

## 3. Condicionales `if`, `elif` y `else`

Los condicionales permiten tomar decisiones para un valor individual. Las funciones que definamos después podrán aplicarse a columnas mediante `Series.apply`, aunque para condiciones simples es preferible la vectorización.

````python
nota = 3.6

if nota >= 4.5:
    nivel = "Alto"
elif nota >= 3.0:
    nivel = "Aprobado"
else:
    nivel = "Debe mejorar"

print(f"Nota {nota}: {nivel}")
````

### 3.1 Clasificar una columna con una función

Aquí usamos `def` para expresar una regla de negocio legible y `apply` para aplicarla a cada valor.

````python
def clasificar_nota(nota):
    if nota >= 4.5:
        return "Alto"
    elif nota >= 3.0:
        return "Aprobado"
    else:
        return "Debe mejorar"

estudiantes["nivel"] = estudiantes["nota"].apply(clasificar_nota)
display(estudiantes)
````

### 3.2 Alternativa vectorizada con NumPy

Para reglas simples sobre toda una columna, `np.where` puede ser más directo que recorrer fila por fila.

````python
estudiantes["estado"] = np.where(
    estudiantes["nota"] >= 3.0,
    "Aprueba",
    "No aprueba"
)
display(estudiantes)
print("Cantidad que aprueba:", (estudiantes["nota"] >= 3.0).sum())
print("Porcentaje que aprueba:", round(100 * (estudiantes["nota"] >= 3.0).mean(), 1), "%")
````

## 4. Pandas y teoría de números: divisibilidad

El operador módulo `%` calcula el residuo. Si `n % k == 0`, entonces `n` es múltiplo de `k` (para `k` distinto de cero).

````python
numeros = pd.DataFrame({"n": np.arange(1, 31)})
numeros["par"] = numeros["n"] % 2 == 0
numeros["multiplo_3"] = numeros["n"] % 3 == 0
numeros["multiplo_5"] = numeros["n"] % 5 == 0
numeros["multiplo_3_y_5"] = numeros["multiplo_3"] & numeros["multiplo_5"]

display(numeros)
print("Pares:", numeros.loc[numeros["par"], "n"].tolist())
print("Múltiplos de 3 y 5:", numeros.loc[numeros["multiplo_3_y_5"], "n"].tolist())
````

## 5. Ciclo `for`: construir listas de divisores

Aunque Pandas y NumPy permiten vectorizar muchas operaciones, los ciclos son valiosos para aprender la lógica de un algoritmo. La función siguiente busca divisores de un entero positivo.

````python
def divisores_de(n):
    divisores = []
    for d in range(1, n + 1):
        if n % d == 0:
            divisores.append(d)
    return divisores

for n in [12, 18, 25]:
    print(f"Divisores de {n}: {divisores_de(n)}")
````

### 5.1 Enriquecer una tabla con resultados de una función

Aplicamos `divisores_de` a cada número y añadimos la cantidad de divisores como nueva columna.

````python
tabla_divisores = pd.DataFrame({"n": np.arange(1, 21)})
tabla_divisores["divisores"] = tabla_divisores["n"].apply(divisores_de)
tabla_divisores["cantidad_divisores"] = tabla_divisores["divisores"].apply(len)

display(tabla_divisores)
print("Números con exactamente 3 divisores:")
display(tabla_divisores.loc[tabla_divisores["cantidad_divisores"] == 3])
````

**Observación matemática:** los enteros positivos que tienen exactamente tres divisores son los cuadrados de números primos.

## 6. Ciclo `while`: detectar números primos

Un número primo es un entero mayor que 1 que tiene exactamente dos divisores positivos: 1 y él mismo. Basta probar posibles divisores hasta que `d * d > n`.

````python
def es_primo(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

for n in [1, 2, 9, 17, 21, 29]:
    print(n, "es primo:", es_primo(n))
````

### 6.1 Crear una columna booleana de primalidad

La función `es_primo` usa un ciclo `while` por dentro. Luego Pandas la aplica a cada número de la tabla.

````python
primos_df = pd.DataFrame({"n": np.arange(1, 101)})
primos_df["es_primo"] = primos_df["n"].apply(es_primo)

display(primos_df.head(15))
print("Primos entre 1 y 100:")
print(primos_df.loc[primos_df["es_primo"], "n"].tolist())
print("Cantidad de primos:", primos_df["es_primo"].sum())
````

## 7. Funciones `def` para preparar datos

Una función puede convertir un valor crudo en una categoría, validar una entrada o producir una característica derivada. La siguiente función clasifica números según su paridad y primalidad.

````python
def etiqueta_numero(n):
    if n < 2:
        return "menor que 2"
    elif es_primo(n):
        return "primo"
    elif n % 2 == 0:
        return "compuesto par"
    else:
        return "compuesto impar"

etiquetas = pd.DataFrame({"n": np.arange(1, 21)})
etiquetas["clasificacion"] = etiquetas["n"].apply(etiqueta_numero)
display(etiquetas)
````

## 8. Funciones `lambda` en Pandas

`lambda` es útil para expresar transformaciones pequeñas. En Pandas puede utilizarse con `apply`, aunque una operación vectorizada suele ser preferible cuando está disponible.

````python
muestra = pd.DataFrame({"n": [12, 25, 37, 40, 58, 63, 75]})
muestra["cuadrado"] = muestra["n"].apply(lambda x: x**2)
muestra["ultimo_digito"] = muestra["n"].apply(lambda x: x % 10)
muestra["termina_en_5"] = muestra["n"].apply(lambda x: x % 10 == 5)

display(muestra)
print("Números que terminan en 5:", muestra.loc[muestra["termina_en_5"], "n"].tolist())
````

### 8.1 Ordenar con `lambda`

El parámetro `key` de `sorted` recibe una función que determina el criterio de ordenamiento.

````python
valores = [31, 14, 22, 7, 45, 18, 29]
ordenados = sorted(valores, key=lambda x: x % 10)
print("Originales:", valores)
print("Ordenados por último dígito:", ordenados)
````

## 9. NumPy y Pandas en un flujo de ciencia de datos

Simularemos mediciones de un sensor con NumPy, construiremos una tabla con Pandas y analizaremos valores atípicos mediante filtros y resúmenes estadísticos.

````python
rng = np.random.default_rng(2026)
mediciones = rng.normal(loc=20, scale=2.5, size=100)
mediciones = np.round(mediciones, 2)
# Incluimos deliberadamente dos valores extremos para practicar su detección.
mediciones[5] = 35.0
mediciones[72] = 8.0

sensor = pd.DataFrame({
    "observacion": np.arange(1, len(mediciones) + 1),
    "medicion": mediciones
})

display(sensor.head())
display(sensor.describe())
````

### 9.1 Filtrar y resumir mediciones

En este ejemplo se considera inusual una medición que está por encima de 25 o por debajo de 12. El umbral es didáctico, no una regla universal.

````python
sensor["es_atipica"] = (sensor["medicion"] > 25) | (sensor["medicion"] < 12)
atipicas = sensor.loc[sensor["es_atipica"]]

print("Número de observaciones:", len(sensor))
print("Cantidad de observaciones atípicas:", len(atipicas))
display(atipicas)
print("Promedio general:", round(sensor["medicion"].mean(), 2))
print("Promedio sin observaciones atípicas:", round(sensor.loc[~sensor["es_atipica"], "medicion"].mean(), 2))
````

### 9.2 Clasificar mediciones con `if`, `elif` y `else`

````python
def clasificar_medicion(x):
    if x < 18:
        return "baja"
    elif x <= 23:
        return "habitual"
    else:
        return "alta"

sensor["categoria"] = sensor["medicion"].apply(clasificar_medicion)
display(sensor.head(10))
print(sensor["categoria"].value_counts())
````

## 10. Agrupación: propiedades de números

Pandas permite agrupar filas y resumir propiedades. Vamos a clasificar enteros por la cantidad de divisores positivos.

````python
def contar_divisores(n):
    cantidad = 0
    for d in range(1, n + 1):
        if n % d == 0:
            cantidad += 1
    return cantidad

numeros_df = pd.DataFrame({"n": np.arange(1, 101)})
numeros_df["es_par"] = numeros_df["n"] % 2 == 0
numeros_df["es_primo"] = numeros_df["n"].apply(es_primo)
numeros_df["cantidad_divisores"] = numeros_df["n"].apply(contar_divisores)
numeros_df["tipo_paridad"] = np.where(numeros_df["es_par"], "par", "impar")

display(numeros_df.head(12))
resumen = numeros_df.groupby("tipo_paridad").agg(
    cantidad=("n", "count"),
    promedio_divisores=("cantidad_divisores", "mean"),
    maximo_divisores=("cantidad_divisores", "max"),
    cantidad_primos=("es_primo", "sum")
)
display(resumen)
````
