## 4. Ciclo `for`: construir una tabla de divisores

Un divisor positivo de `n` es un entero positivo `d` que divide a `n` sin dejar residuo. Aquí combinamos un ciclo con condicionales y luego guardamos resultados en listas, que convertimos en arreglos NumPy.

````python
n = 24
divisores = []

for d in range(1, n + 1):
    if n % d == 0:
        divisores.append(d)

divisores = np.array(divisores)
print("Divisores de", n, ":", divisores)
print("Cantidad de divisores:", divisores.size)
````

### 4.1 Clasificar números con un ciclo

El siguiente ejemplo recorre números del 1 al 15 y clasifica cada uno según su divisibilidad.

````python
for n in range(1, 16):
    if n % 3 == 0 and n % 5 == 0:
        print(n, "-> múltiplo de 3 y 5")
    elif n % 3 == 0:
        print(n, "-> múltiplo de 3")
    elif n % 5 == 0:
        print(n, "-> múltiplo de 5")
    else:
        print(n, "-> ninguno de los dos")
````

## 5. Ciclo `while`: encontrar un divisor y probar primalidad

Un número primo es un entero mayor que 1 cuyos únicos divisores positivos son 1 y él mismo. El ciclo `while` es útil cuando queremos repetir una búsqueda mientras una condición se mantenga verdadera.

````python
n = 13
d = 2

while d < n and n % d != 0:
    d += 1

if n > 1 and d < n:
    print(n, "no es primo; un divisor propio es", d)
elif n > 1:
    print(n, "es primo")
else:
    print(n, "no es primo porque es menor que 2")
````

### 5.1 Una función de primalidad más eficiente

No hace falta probar todos los divisores hasta `n - 1`: si `n` es compuesto, tiene algún divisor no mayor que √`n`. `int(n**0.5)` calcula la parte entera de la raíz cuadrada.

````python
def es_primo(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

for n in range(1, 21):
    if es_primo(n):
        print(n, "es primo")
````

## 6. Funciones creadas con `def`

Las funciones permiten reutilizar código, nombrar una operación y probarla por separado.

````python
def es_multiplo(n, k):
    """Devuelve True si n es múltiplo de k; k debe ser distinto de cero."""
    if k == 0:
        raise ValueError("k no puede ser cero")
    return n % k == 0

print(es_multiplo(21, 7))
print(es_multiplo(22, 7))
````

### 6.1 Aplicar una función a datos de NumPy

Una función escalar como `es_primo` recibe un número. Para evaluarla sobre varios valores, podemos recorrer el arreglo con `for` y guardar los resultados.

````python
datos = np.arange(1, 31)
mascara_primos = np.array([es_primo(int(n)) for n in datos])

print("Datos:", datos)
print("Máscara de primos:", mascara_primos)
print("Primos:", datos[mascara_primos])
print("Cantidad de primos:", np.sum(mascara_primos))
````

## 7. Funciones `lambda`

Una expresión `lambda` crea una función pequeña de una sola expresión. Es útil para operaciones breves; para algoritmos con ciclos y varias decisiones, normalmente conviene `def`.

````python
cuadrado = lambda x: x**2
es_par = lambda x: x % 2 == 0
ultimo_digito = lambda x: x % 10

print("Cuadrado de 8:", cuadrado(8))
print("¿42 es par?", es_par(42))
print("Último dígito de 137:", ultimo_digito(137))
````

### 7.1 Ordenar y transformar valores

````python
valores = [31, 14, 22, 7, 45, 18, 29]
ordenados_por_ultimo_digito = sorted(valores, key=lambda x: x % 10)

print("Originales:", valores)
print("Ordenados por último dígito:", ordenados_por_ultimo_digito)

arreglo = np.array(valores)
print("Cuadrados con NumPy:", arreglo**2)
````

## 8. Ciencia de datos con arreglos: resumen de mediciones

Supongamos que un sensor registra mediciones. Sin Pandas ni Matplotlib, podemos calcular estadísticas básicas, detectar valores fuera de un umbral y clasificar las observaciones.

## 9. Conectar teoría de números y ciencia de datos

Vamos a analizar un conjunto de identificadores enteros. Para cada identificador nos interesa su paridad, si es primo y su cantidad de divisores. La función siguiente combina condicionales y ciclos.

````python
def contar_divisores(n):
    if n < 1:
        return 0
    cantidad = 0
    for d in range(1, n + 1):
        if n % d == 0:
            cantidad += 1
    return cantidad

identificadores = np.arange(1, 21)
es_primo_vector = np.array([es_primo(int(n)) for n in identificadores])
cantidad_divisores = np.array([contar_divisores(int(n)) for n in identificadores])
es_par_vector = identificadores % 2 == 0

print("Identificadores:", identificadores)
print("Pares:", identificadores[es_par_vector])
print("Primos:", identificadores[es_primo_vector])
print("Con exactamente 4 divisores:", identificadores[cantidad_divisores == 4])
````

# Solución: Clasificación de outliers con la regla IQR

## Enunciado
Dado un array de NumPy con datos de ventas, clasificar cada valor usando la regla del rango intercuartílico (IQR) y retornar una lista con las etiquetas.

---



## Explicación de la regla IQR

La regla del **rango intercuartílico (IQR)** es un método estándar en ciencia de datos para detectar valores atípicos:

1. **Q1** = primer cuartil (percentil 25)
2. **Q3** = tercer cuartil (percentil 75)
3. **IQR** = Q3 − Q1
4. **Límite inferior** = Q1 − 1.5 × IQR
5. **Límite superior** = Q3 + 1.5 × IQR

Cualquier valor **fuera** de `[límite_inferior, límite_superior]` es un outlier.

---



## Solución con `def` (recomendada para enseñar)



````python
import numpy as np

def clasificar_outliers(datos):
    """
    Clasifica cada valor de un array según la regla del IQR.
    Retorna una lista de etiquetas: 'outlier inferior', 'outlier superior' o 'normal'.
    """
    # 1. Calcular Q1, Q3 e IQR
    q1 = np.percentile(datos, 25)
    q3 = np.percentile(datos, 75)
    iqr = q3 - q1

    # 2. Calcular los límites (cercas de Tukey)
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    # 3. Clasificar cada valor
    etiquetas = []
    for x in datos:
        if x < limite_inferior:
            etiquetas.append("outlier inferior")
        elif x > limite_superior:
            etiquetas.append("outlier superior")
        else:
            etiquetas.append("normal")

    return etiquetas


# --- Datos del problema ---
datos = np.array([10, 12, 14, 15, 16, 18, 20, 22, 100])

# --- Ejecución ---
resultado = clasificar_outliers(datos)
print(resultado)
````


## Verificación paso a paso

## Versión vectorizada con NumPy (bonus)

Para mostrar a los estudiantes cómo NumPy permite evitar ciclos explícitos:



````python
import numpy as np

def clasificar_outliers_vectorizado(datos):
    q1 = np.percentile(datos, 25)
    q3 = np.percentile(datos, 75)
    iqr = q3 - q1
    li = q1 - 1.5 * iqr
    ls = q3 + 1.5 * iqr

    # np.select aplica condiciones vectorizadas
    condiciones = [datos < li, datos > ls]
    etiquetas = ["outlier inferior", "outlier superior"]
    return np.select(condiciones, etiquetas, default="normal").tolist()


print(clasificar_outliers_vectorizado(datos))
# ['normal', 'normal', 'normal', 'normal', 'normal', 'normal', 'normal', 'normal', 'outlier superior']
````
